#1. Perguntas e objetivos
Para este projeto foi dividido em novas pequenas perguntas que podem ser respondidas com os dados públicos coletados, e com isso, chegar em uma breve conclusão sobre alguns dos problemas que o nosso planeta enfrenta a como entender e ajudar de alguma forma e melhorar nossa qualidade de vida (geral).

1. Qual foi a emissão de carbono no planeta por ano?
2. Qual foi a emissão de carbono no planeta por década?
3. Emissão de carbono no planeta por país.
4. Qual foi a emissão de carbono no planeta por país por ano?
5. Qual foi a emissão de carbono no planeta por país por década?
6. Emissão de carbono no planeta por continente.
7. Qual foi a emissão de carbono no planeta por continente por ano?
8. Qual foi a emissão de carbono no planeta por continente por década?
9. Sobre alguns dados coletados estarem negativos.
10. Quais são os países que possuem emissão negativa?
11. Quais são os continentes que possuem emissão negativa?

##4.1 Conversão de arquivos NC e Junção.
Este é um script para ler e converter os arquivos de dados que possuem o formato "NC", e após finalizar a conversão desses arquivos o script irá combinar os arquivos para o formato CSV que estiverem com o mesmo nome.

```python
import xarray as xr
import pandas as pd
import os

# Caminho para a pasta contendo os arquivos .nc
data_path = "D:/dados_nc"  # Altere para o caminho correto da sua pasta

# Listar os arquivos .nc na pasta
file_paths = [os.path.join(data_path, file) for file in os.listdir(data_path) if file.endswith(".nc")]

# Verificar se há arquivos .nc
if not file_paths:
    raise ValueError("Nenhum arquivo .nc encontrado na pasta.")

# Carregar cada arquivo individualmente e garantir que os índices de tempo sejam monotônicos
datasets = []
for file_path in file_paths:
    try:
        ds = xr.open_dataset(file_path, decode_times=False)
        
        # Garantir que a dimensão 'time' esteja monotônica
        if "time" in ds.variables:
            ds = ds.sortby("time")  # Ordenar por 'time'
        
        datasets.append(ds)
    except Exception as e:
        print(f"Falha ao carregar o arquivo {file_path}: {e}")

# Concatenar os datasets manualmente
try:
    combined_data = xr.concat(datasets, dim="time")  # Usar 'time' como dimensão de concatenação
    print("Arquivos combinados com sucesso!")
except Exception as e:
    raise ValueError(f"Falha ao combinar arquivos: {e}")

# Decodificar manualmente a variável 'time' usando cftime
if "time" in combined_data.variables:
    time_units = combined_data["time"].attrs.get("units", "")
    raw_time_values = combined_data["time"].values  # Valores brutos (números)

    # Identificar o tipo de unidade de tempo
    if "days since" in time_units:
        reference_date = pd.Timestamp(time_units.split("days since ")[1])  # Extrair a data de referência
        decoded_dates = [reference_date + pd.Timedelta(days=int(t)) for t in raw_time_values]
    elif "months since" in time_units:
        reference_date = pd.Timestamp(time_units.split("months since ")[1])  # Extrair a data de referência
        decoded_dates = [reference_date + pd.DateOffset(months=int(t)) for t in raw_time_values]
    else:
        raise ValueError(f"Unidades de tempo não suportadas: {time_units}")

    # Atualizar a variável 'time' no dataset
    combined_data["time"] = xr.Variable(
        dims=combined_data["time"].dims,
        data=decoded_dates,
        attrs=combined_data["time"].attrs
    )

# Converter para DataFrame
df = combined_data.to_dataframe()

# Salvar como CSV
output_path = os.path.join(data_path, "dados_combinados.csv")
df.to_csv(output_path)

print(f"Arquivos combinados e salvos como CSV em: {output_path}")
```

##4.2 Separação de arquivos CSV
Há alguns arquivos que após a conversão do formato NC para CSV, que ficaram com um tamanho muito grande, e então não puderam ser importados no GitHub, então foi criado esse script para separá-los em arquivos de 50 mb no máximo e enviar para o GitHub. 

```python
import pandas as pd
import os

def split_csv_by_size(input_file, output_dir, max_size_mb=50):
    """
    Separa um arquivo CSV grande em arquivos menores com base no tamanho máximo.

    :param input_file: Caminho para o arquivo CSV de entrada.
    :param output_dir: Diretório onde os arquivos divididos serão salvos.
    :param max_size_mb: Tamanho máximo de cada arquivo dividido (em MB).
    """
    # Converte o tamanho máximo para bytes
    max_size_bytes = max_size_mb * 1024 * 1024

    # Cria o diretório de saída, se ele não existir
    os.makedirs(output_dir, exist_ok=True)

    # Lê o cabeçalho do arquivo CSV
    with open(input_file, 'r') as f:
        header = f.readline()

    # Inicializa variáveis
    chunk_size = 10000  # Número inicial de linhas por chunk (ajustável)
    file_index = 1
    current_chunk = []
    current_size = len(header.encode('utf-8'))  # Começa com o tamanho do cabeçalho

    # Lê o arquivo CSV em chunks
    for chunk in pd.read_csv(input_file, chunksize=chunk_size):
        for _, row in chunk.iterrows():
            # Converte a linha para string e calcula seu tamanho em bytes
            row_str = ','.join(map(str, row.values)) + '\n'
            row_size = len(row_str.encode('utf-8'))

            # Verifica se a linha cabe no arquivo atual
            if current_size + row_size > max_size_bytes:
                # Salva o chunk atual como um novo arquivo
                output_file = os.path.join(output_dir, f'part_{file_index}.csv')
                with open(output_file, 'w') as f:
                    f.write(header)
                    f.writelines(current_chunk)
                
                # Reinicia variáveis para o próximo arquivo
                file_index += 1
                current_chunk = [row_str]
                current_size = len(header.encode('utf-8')) + row_size
            else:
                # Adiciona a linha ao chunk atual
                current_chunk.append(row_str)
                current_size += row_size

    # Salva o último chunk, se houver
    if current_chunk:
        output_file = os.path.join(output_dir, f'part_{file_index}.csv')
        with open(output_file, 'w') as f:
            f.write(header)
            f.writelines(current_chunk)

    print(f"Arquivo dividido em {file_index} partes.")

# Exemplo de uso
input_csv = "D:/dados_nc/.csv"
output_directory = "caminho/para/diretorio_saida"
split_csv_by_size(input_csv, output_directory, max_size_mb=50)
```

```python
%sql
-- SQL CARBON_MONITOR - HISTORICAL EMISSIONS
SELECT COUNTRY PAIS, 
       CAST(substr(DATE, 7, 10) AS NUMERIC) ANO,
       SECTOR SETOR,
       SUM(VALUE) VALOR --CO2 emitidos por ano em metros 
FROM CARBON_MONITOR
GROUP BY 
       COUNTRY, 
       CAST(substr(DATE, 7, 10) AS NUMERIC),
       SECTOR
       ; 

--SQL CLIMATE WATCH AGRICULTURE EMISSION 
WITH EMISSIONS AS (
       SELECT 
              AREA PAIS, 
              YEAR ANO, 
              `short names` PRODUTO,
              EMISSIONS VALOR --CO2  kg CO2eq/kg de produto 
       FROM cw_agriculture_emissions
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1961`, `1962`, `1963`, `1964`, `1965`, `1966`, `1967`, `1968`, `1969`, `1970`,
              `1971`, `1972`, `1973`, `1974`, `1975`, `1976`, `1977`, `1978`, `1979`, `1980`,
              `1981`, `1982`, `1983`, `1984`, `1985`, `1986`, `1987`, `1988`, `1989`, `1990`,
              `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`,
              `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`,
              `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`,
              `2021`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;

--SQL CLIMATE WATCH - HISTORICAL EMISSIONS
WITH EMISSIONS AS (
       SELECT 
              ISO PAIS, 
              YEAR ANO, 
              SECTOR SETOR,
              EMISSIONS VALOR --CO2  MtCO₂e
       FROM cw_historical_emissions
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1961`, `1962`, `1963`, `1964`, `1965`, `1966`, `1967`, `1968`, `1969`, `1970`,
              `1971`, `1972`, `1973`, `1974`, `1975`, `1976`, `1977`, `1978`, `1979`, `1980`,
              `1981`, `1982`, `1983`, `1984`, `1985`, `1986`, `1987`, `1988`, `1989`, `1990`,
              `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`,
              `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`,
              `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`,
              `2021`, `2022`, `2023`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;

--FAOSTAT - HISTORICAL EMISSIONS TOTALS
WITH EMISSIONS AS (
       SELECT 
              AREA PAIS, 
              substr(YEAR, 2, 4) ANO, 
              ITEM PRODUTO,
              ELEMENT GAS,
              EMISSIONS VALOR --CO2  KT (kt)
       FROM faostat_emissions_totals
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `Y1961`, `Y1962`, `Y1963`, `Y1964`, `Y1965`, `Y1966`, `Y1967`, `Y1968`, `Y1969`, `Y1970`,
              `Y1971`, `Y1972`, `Y1973`, `Y1974`, `Y1975`, `Y1976`, `Y1977`, `Y1978`, `Y1979`, `Y1980`,
              `Y1981`, `Y1982`, `Y1983`, `Y1984`, `Y1985`, `Y1986`, `Y1987`, `Y1988`, `Y1989`, `Y1990`,
              `Y1991`, `Y1992`, `Y1993`, `Y1994`, `Y1995`, `Y1996`, `Y1997`, `Y1998`, `Y1999`, `Y2000`,
              `Y2001`, `Y2002`, `Y2003`, `Y2004`, `Y2005`, `Y2006`, `Y2007`, `Y2008`, `Y2009`, `Y2010`,
              `Y2011`, `Y2012`, `Y2013`, `Y2014`, `Y2015`, `Y2016`, `Y2017`, `Y2018`, `Y2019`, `Y2020`,
              `Y2021`, `Y2022`, `Y2023`, `Y2030`, `Y2050` 
       )
       ) AS unpivoted_data
       WHERE `Element Code` IN ('724313','723113','724413','7273','717815')
)
SELECT B.*
  FROM EMISSIONS B;


--IEA - EMISSIONS PER COUNTRIES 
WITH EMISSIONS AS (
       SELECT 
              Country PAIS, 
              YEAR ANO,  
              `Type emission` GAS,
              EMISSIONS VALOR, --CO2  Mt CO2
              `Grow rate % 2019-20` COMPARATIVO_2019_20_PERC,
              `Grow rate % 2020-21` COMPARATIVO_2020_21_PERC
       FROM iea_emissions
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `2019`, `2020`, `2021`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;

  
--IEA - EMISSIONS PER  GASES EMISSIONS
WITH EMISSIONS AS (
       SELECT 
              YEAR ANO,  
              `Type emission` GAS,
              `CO2 emissions` PRODUTO,
              EMISSIONS VALOR, --CO2  Mt CO2
              `Grow rate % 2019-20` COMPARATIVO_2019_20_PERC,
              `Grow rate % 2020-21` COMPARATIVO_2020_21_PERC
       FROM iea_gas_emissions
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `2019`, `2020`, `2021`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;


--NASA - CO2 FLUX TOTAL  
SELECT ANO,
       LATITUDE_LONGITUDE,
       SUM(VALOR) VALOR,
       qtd_area_m2
FROM (
       SELECT DISTINCT 
              CAST(substr(CAST(TRUNC(TIME, 'YEAR') AS STRING), 1, 4) AS NUMERIC) ANO,
              latitude || ',' || longitude LATITUDE_LONGITUDE,
              total_post VALOR, --g/m^2/day
              area qtd_area_m2
       FROM nasa_cms_flux_total
)
GROUP BY ANO,
       LATITUDE_LONGITUDE,
       qtd_area_m2
ORDER BY 1, 2;


--ONU - HISTORICAL EMISSIONS
WITH EMISSIONS AS (
       SELECT 
              PARTY PAIS,
              YEAR ANO,  
              EMISSIONS VALOR, --KT
              `Base Year` VALOR_ANO_BASE, --kt
              `Change from base year to latest reported year` COMPARATIVO_ANO_BASE_ULTIMO_ANO, --%
              `Last Inventory Year 2021` ANO_ULTIMO_INVENTARIO --%      
       FROM onu_time_series_co_2_total
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1990`, `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`, `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`, `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;

  --ONU - CO2 REMOVIDOS
WITH EMISSIONS AS (
       SELECT 
              PARTY PAIS,
              YEAR ANO,  
              EMISSIONS VALOR, --KT
              `Base Year` VALOR_ANO_BASE, --kt
              `Last Inventory Year (2021)` ANO_ULTIMO_INVENTARIO --%      
       FROM onu_annual_emissions_removals
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1990`, `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`, `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`, `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;

  --ONU - GREENHOUSE GAS
WITH EMISSIONS AS (
       SELECT 
              Category SECTOR,
              YEAR ANO,  
              EMISSIONS VALOR, --KT
              `Base year` VALOR_ANO_BASE,
              `Last Inventory Year 2021` ANO_ULTIMO_INVENTARIO
       FROM onu_annual_greenhouse_gas
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1990`, `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`, `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`, `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;


--OUR WORLD DATA - CO2 EMISSIONS ANUAL PER COUNTRY      
SELECT CODE PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       ENTITY NOME_PAIS, 
       YEAR ANO,       
       SUM(`Annual CO₂ emissions`) VALOR
  FROM workspace.pucrio.owd_annual_co_2_emissions_per_country
 GROUP BY CODE,
       YEAR,
       Entity;

--OUR WORLD DATA -- CO2 EMISSIONS SHARE ANNUAL CO2
SELECT CODE PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       ENTITY NOME_PAIS, 
       YEAR ANO,       
       SUM(`Share of global annual CO₂ emissions`) VALOR
  FROM workspace.pucrio.owd_annual_share_of_co_2_emissions
 GROUP BY CODE,
       YEAR,
       Entity;       


--OUR WORLD DATA -- CO2 EMISSIONS FACTOR
SELECT ENTITY NOME_FATOR, 
       YEAR ANO,       
       SUM(`CO₂ emission factor`) VALOR
  FROM  workspace.pucrio.owd_carbon_dioxide_emissions_factor
 GROUP BY 
       YEAR,
       Entity;       


--OUR WORLD DATA -- CO2 CONCENTRATION LONG TERM (ONLY WORLD)
SELECT YEAR ANO, --HÁ ANOS NEGATIVOS E ANOS COM VALORES MUITO BAIXOS
       SUM(`Annual concentration of atmospheric carbon dioxide`) VALOR
  FROM workspace.pucrio.owd_co_2_long_term_concentration
 WHERE YEAR > 1950
 GROUP BY YEAR;       



--OUR WORLD DATA -- CO2 EMISSIONS PER CAPITA
SELECT Entity NOME_PAIS, 
       Code PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       YEAR ANO,
       SUM(`Annual CO₂ emissions (per capita)`) VALOR
  FROM  workspace.pucrio.owd_co_emissions_per_capita 
 GROUP BY Entity, 
       Code, 
       YEAR;           



--OUR WORLD DATA -- CO2 EMISSIONS CUMALITIVE
SELECT Entity NOME_PAIS, 
       Code PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       YEAR ANO,
       SUM(`Cumulative CO₂ emissions`) VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions
 GROUP BY Entity, 
       Code, 
       YEAR;                



--OUR WORLD DATA -- CO2 CUMALITIVE SHARE 
SELECT Entity NOME_PAIS, 
       Code PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       YEAR ANO,
       SUM(`Share of global cumulative CO₂ emissions`) VALOR
  FROM   workspace.pucrio.owd_share_of_cumulative_co_2
 GROUP BY Entity, 
       Code, 
       YEAR;       


--OUR WORLD DATA -- CO2 EMISSIONS FOR ALL 
SELECT 
  country AS PAIS,
  ISO_CODE AS SIGLA_PAIS,
  year AS ANO,
  population AS POPULACAO,
  gdp AS PIB,

  --  CIMENTO
  SUM(cement_co2) AS VALOR_EMITIDO_CO2_CIMENTO,
  cement_co2_per_capita AS VALOR_PER_CAPITA_CO2_CIMENTO,

  --  CARVÃO
  SUM(coal_co2) AS VALOR_EMITIDO_CO2_CARBONIFERO,
  coal_co2_per_capita AS VALOR_PER_CAPITA_CO2_CARBONIFERO,

  --  GÁS NATURAL
  SUM(gas_co2) AS VALOR_EMITIDO_CO2_GAS_NATURAL,
  gas_co2_per_capita AS VALOR_PER_CAPITA_CO2_GAS_NATURAL,

  --  PETRÓLEO
  SUM(oil_co2) AS VALOR_EMITIDO_CO2_PETROLEO,
  oil_co2_per_capita AS VALOR_PER_CAPITA_CO2_PETROLEO,

  --  QUEIMA DE GAS FLARE
  SUM(flaring_co2) AS VALOR_EMITIDO_CO2_REVEJAMENTO,
  flaring_co2_per_capita AS VALOR_PER_CAPITA_CO2_REVEJAMENTO,

  --  MUDANÇA DE USO DA TERRA
  SUM(land_use_change_co2) AS VALOR_EMITIDO_CO2_MUDANCA_USO_SOLO,
  land_use_change_co2_per_capita AS VALOR_PER_CAPITA_CO2_MUDANCA_USO_SOLO,

  --  OUTRAS INDÚSTRIAS
  SUM(other_industry_co2) AS VALOR_EMITIDO_CO2_OUTRAS_INDUSTRIAS,

  -- CONSUMO E COMÉRCIO
  SUM(consumption_co2) AS VALOR_EMITIDO_CO2_CONSUMO,
  consumption_co2_per_capita AS VALOR_PER_CAPITA_CO2_CONSUMO,
  consumption_co2_per_gdp AS VALOR_PER_PIB_CO2_CONSUMO,
  SUM(trade_co2) AS VALOR_EMITIDO_CO2_COMERCIO,
  trade_co2_share AS PARTICIPACAO_CO2_COMERCIO_NO_GLOBAL,

  --  CO₂ TOTAL
  SUM(co2) AS VALOR_EMITIDO_CO2,
  co2_per_capita AS VALOR_PER_CAPITA_CO2,
  co2_per_gdp AS VALOR_PER_PIB_CO2,
  co2_per_unit_energy AS VALOR_PER_UNIDADE_ENERGIA_CO2,
  co2_growth_abs AS CRESCIMENTO_ABSOLUTO_CO2,
  co2_growth_prct AS CRESCIMENTO_PERCENTUAL_CO2,

  --  CO₂ INCLUINDO MUDANÇA DE USO DA TERRA
  SUM(co2_including_luc) AS VALOR_EMITIDO_CO2_INCL_MUDANCA_USO_SOLO,
  co2_including_luc_per_capita AS VALOR_PER_CAPITA_CO2_INCL_MUDANCA_USO_SOLO,
  co2_including_luc_per_gdp AS VALOR_PER_PIB_CO2_INCL_MUDANCA_USO_SOLO,
  co2_including_luc_per_unit_energy AS VALOR_PER_UNIDADE_ENERGIA_CO2_INCL_MUDANCA_USO_SOLO,
  co2_including_luc_growth_abs AS CRESCIMENTO_ABSOLUTO_CO2_INCL_MUDANCA_USO_SOLO,
  co2_including_luc_growth_prct AS CRESCIMENTO_PERCENTUAL_CO2_INCL_MUDANCA_USO_SOLO,

  --  METANO (CH4)
  SUM(methane) AS VALOR_EMITIDO_METANO,
  methane_per_capita AS VALOR_PER_CAPITA_METANO,

  --  ÓXIDO NITROSO (N2O)
  SUM(nitrous_oxide) AS VALOR_EMITIDO_OXIDO_NITROSO,
  nitrous_oxide_per_capita AS VALOR_PER_CAPITA_OXIDO_NITROSO,

  --  GASES DO EFEITO ESTUFA (GHG)
  SUM(total_ghg) AS VALOR_EMITIDO_TODOS_GASES_ESTUFA,
  SUM(total_ghg_excluding_lucf) AS VALOR_EMITIDO_GEE_EXC_LUCF,
  ghg_per_capita AS VALOR_PER_CAPITA_GEE,
  ghg_excluding_lucf_per_capita AS VALOR_PER_CAPITA_GEE_EXC_LUCF,

  --  ENERGIA
  primary_energy_consumption AS CONSUMO_PRIMARIO_ENERGIA,
  energy_per_capita AS ENERGIA_PER_CAPITA,
  energy_per_gdp AS ENERGIA_PER_PIB,

  --  PARTICIPAÇÃO GLOBAL (%)
  share_global_cement_co2 AS PARTICIPACAO_GLOBAL_CO2_CIMENTO,
  share_global_co2 AS PARTICIPACAO_GLOBAL_CO2,
  share_global_co2_including_luc AS PARTICIPACAO_GLOBAL_CO2_INCL_MUDANCA_USO_SOLO,
  share_global_coal_co2 AS PARTICIPACAO_GLOBAL_CO2_CARBONIFERO,
  share_global_gas_co2 AS PARTICIPACAO_GLOBAL_CO2_GAS_NATURAL,
  share_global_oil_co2 AS PARTICIPACAO_GLOBAL_CO2_PETROLEO,
  share_global_flaring_co2 AS PARTICIPACAO_GLOBAL_CO2_REVEJAMENTO,
  share_global_other_co2 AS PARTICIPACAO_GLOBAL_OUTROS_CO2,

  --  PARTICIPAÇÃO ACUMULADA GLOBAL
  cumulative_cement_co2 AS TOTAL_ACUMULADO_CO2_CIMENTO,
  cumulative_coal_co2 AS TOTAL_ACUMULADO_CO2_CARBONIFERO,
  cumulative_gas_co2 AS TOTAL_ACUMULADO_CO2_GAS_NATURAL,
  cumulative_oil_co2 AS TOTAL_ACUMULADO_CO2_PETROLEO,
  cumulative_flaring_co2 AS TOTAL_ACUMULADO_CO2_REVEJAMENTO,
  cumulative_luc_co2 AS TOTAL_ACUMULADO_CO2_MUDANCA_USO_SOLO,
  cumulative_other_co2 AS TOTAL_ACUMULADO_OUTROS_CO2,
  cumulative_co2 AS TOTAL_ACUMULADO_CO2,
  cumulative_co2_including_luc AS TOTAL_ACUMULADO_CO2_INCL_MUDANCA_USO_SOLO,

  --  PARTICIPAÇÃO EM EMISSÕES ACUMULADAS
  share_global_cumulative_cement_co2 AS PARTICIPACAO_ACUMULADA_CO2_CIMENTO,
  share_global_cumulative_coal_co2 AS PARTICIPACAO_ACUMULADA_CO2_CARBONIFERO,
  share_global_cumulative_gas_co2 AS PARTICIPACAO_ACUMULADA_CO2_GAS_NATURAL,
  share_global_cumulative_oil_co2 AS PARTICIPACAO_ACUMULADA_CO2_PETROLEO,
  share_global_cumulative_flaring_co2 AS PARTICIPACAO_ACUMULADA_CO2_REVEJAMENTO,
  share_global_cumulative_luc_co2 AS PARTICIPACAO_ACUMULADA_CO2_MUDANCA_USO_SOLO,
  share_global_cumulative_other_co2 AS PARTICIPACAO_ACUMULADA_OUTROS_CO2,
  share_global_cumulative_co2 AS PARTICIPACAO_ACUMULADA_CO2,
  share_global_cumulative_co2_including_luc AS PARTICIPACAO_ACUMULADA_CO2_INCL_MUDANCA_USO_SOLO,

  -- MUDANÇA DE TEMPERATURA (°C)
  temperature_change_from_co2 AS MUDANCA_TEMPERATURA_DEVIDO_CO2,
  temperature_change_from_ch4 AS MUDANCA_TEMPERATURA_DEVIDO_METANO,
  temperature_change_from_n2o AS MUDANCA_TEMPERATURA_DEVIDO_OXIDO_NITROSO,
  temperature_change_from_ghg AS MUDANCA_TEMPERATURA_DEVIDO_GEE,
  share_of_temperature_change_from_ghg AS PARTICIPACAO_MUDANCA_TEMP_GEE

FROM  workspace.pucrio.owid_co_2_data-- Substitua pelo nome real da sua tabela
WHERE YEAR > 1950
GROUP BY 
  country, ISO_CODE, year, population, gdp,
  cement_co2_per_capita,
  coal_co2_per_capita,
  gas_co2_per_capita,
  oil_co2_per_capita,
  flaring_co2_per_capita,
  land_use_change_co2_per_capita,
  consumption_co2_per_capita,
  consumption_co2_per_gdp,
  co2_per_capita,
  co2_per_gdp,
  co2_per_unit_energy,
  co2_growth_abs,
  co2_growth_prct,
  co2_including_luc_per_capita,
  co2_including_luc_per_gdp,
  co2_including_luc_per_unit_energy,
  co2_including_luc_growth_abs,
  co2_including_luc_growth_prct,
  methane_per_capita,
  nitrous_oxide_per_capita,
  ghg_per_capita,
  ghg_excluding_lucf_per_capita,
  primary_energy_consumption,
  energy_per_capita,
  energy_per_gdp,
  trade_co2_share,
  share_global_cement_co2,
  share_global_co2,
  share_global_co2_including_luc,
  share_global_coal_co2,
  share_global_gas_co2,
  share_global_oil_co2,
  share_global_flaring_co2,
  share_global_other_co2,
  share_global_cumulative_cement_co2,
  share_global_cumulative_co2,
  share_global_cumulative_co2_including_luc,
  share_global_cumulative_coal_co2,
  share_global_cumulative_gas_co2,
  share_global_cumulative_oil_co2,
  share_global_cumulative_flaring_co2,
  share_global_cumulative_luc_co2,
  share_global_cumulative_other_co2,
  share_of_temperature_change_from_ghg,
  temperature_change_from_co2,
  temperature_change_from_ch4,
  temperature_change_from_n2o,
  temperature_change_from_ghg,
  cumulative_cement_co2,
   cumulative_coal_co2,
  cumulative_gas_co2 ,
  cumulative_oil_co2,
  cumulative_flaring_co2 ,
  cumulative_luc_co2,
  cumulative_other_co2,
  cumulative_co2 ,
  cumulative_co2_including_luc;


```
