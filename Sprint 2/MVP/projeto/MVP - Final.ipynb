#1. Perguntas e objetivos
Para este projeto foi dividido em novas pequenas perguntas que podem ser respondidas com os dados públicos coletados, e com isso, chegar em uma breve conclusão sobre alguns dos problemas que o nosso planeta enfrenta a como entender e ajudar de alguma forma e melhorar nossa qualidade de vida (geral).

1. Qual a emissão de carbono no planeta por ano?
2. Qual a emissão de carbono no planeta por país por ano?
3. Qual a emissão de carbono no planeta por continente?
4. Qual a emissão de carbono no planeta por continente por ano?
5. Qual a taxa de absorção de dióxido de carbono por ano?
6. Qual a taxa de absorção de dióxido de carbono por continente?
7. Qual a taxa de absorção de dióxido de carbono por país?

#2. Coleta dos dados
As seguintes fontes de dados foram usadas para a coleta dos dados, são elas: Carbon Monitor, Climate Watch (CW), International Energy Agency (IEA), National Aeronautics and Space Administration (NASA), Organização das Nações Unidas (ONU), Our World in Data, World Bank Group, porém nem todas serão usadas, todos os detalhes estarão na sequência.

Todas as organizações listadas anteriormente, possuem dados públicos que foram cuidadosamente baixados, interpretados e verificados para a organização deste projeto. Abaixo os links para dos dados.

##3.2. Climate Watch


Link: https://www.climatewatchdata.org

Arquivo para download da fonte de dados:
https://www.climatewatchdata.org/data-explorer/historical-emissions?historical-emissions-data-sources=climate-watch&historical-emissions-gases=all-ghg&historical-emissions-regions=All%20Selected&historical-emissions-sectors=total-including-lulucf&page=1

Como baixar os arquivos: 
1. Clique em Download Bulk Data
2. Baixe dois arquivos, sendo eles: Agriculture Profile e GHG Emissions
3. Ao clicar nos links das fontes de dados será direcionado para um cadastro de newslatter, bem simples, basta preencher o cadastro e o download será iniciado.
4. Ainda na tela de historical emissions, faça os seguintes filtros para realizar o donwload de todo o histórico de emissões de gás de dióxido de carbono:
- 4.1. Data Sources - All Selected
- 4.2. Parties and Regions - All Selected
- 4.3. Sectors - All Selected
- 4.4. Gases - CO2
- 4.5. Start Year - 1950
- 4.6. End Year - 2023
- 4.7. Bem abaixo do gráfico há um botão chamado: Download Historical Data, clique para realizar o download.

Local no repositório de arquivos coletados:
\projeto\Data\Climate Watch\

1. Há dados com os valores de country (países) com a seguinte informação: ANNEXI, talvez seja o valor do arquivo para o planeta, pois não há essa Contry no arquivo, ao contrário dos demais arquivos, entretanto também não há evidências suficiente para afirmar isso, então, esses campos serão descartados.

3. Há alguns setores que possuem valores agregados para vários tipos de gases, e outros de tipos de pesquisas que estão juntos, como o objetivo deste projeto é apenas de dióxido de carbono, estes valores serão descartados.

Filtros utilizados para obter os dados
1. Countries - Select all
2. Elements - Emissions CO2
3. Items - Select All
4. Years - Select all
5. FAO Source - Select All
6. Ohters Sources - Select All

Para realização do download:
1. Seguir para Bulk Downloads
2. All Data
3. Descompactar os arquivos
4. O arquivo com os dados é o Emissions_Totals_E_All_Data.csv

Observação principal: 
- Será usado o arquivo Emissions_Totals_E_All_Data_NOFLAG.csv, pois o arquivo com FLAGs, há marcações que não estão descritas na documentação, então não sabe-se o que poderá ser estas marcações.

- Os demais arquivos são de códigos e outros, entretanto poderá ser descartado para este projeto.



Documentação dos campos:
1. Area code - código da área - descrito no arquivo de Emissions_Totals_E_AreaCodes.csv
2. Area code M49 - código da área conforme a tabela M49 - descrito no arquivo de Emissions_Totals_E_AreaCodes.csv
3. Area - Nome do país ou da área do registro
4. Item Code - código dos itens
* Era para ter os registros de códigos dos itens no arquivo: Emissions_Totals_E_ItemCodes.csv, porém este arquivo não possuí nenhum informação, apenas o cabeçalho.
5. Item - Nome do item
6. Element Code - Código do elemento do registro e/ou da emissão - disponível no arquivo Emissions_Totals_E_Elements.csv
7. Element - nome do elemento do registro

#4. Scripts auxiliares

##4.1 Conversão de arquivos NC e Junção.
Este é um script para ler e converter os arquivos de dados que possuem o formato "NC", e após finalizar a conversão desses arquivos o script irá combinar os arquivos para o formato CSV que estiverem com o mesmo nome.

```python
pip install xarray
```

```python
%restart_python
```

```python
pip install pandas
```

```python
%restart_python
```

```python
import xarray as xr
import pandas as pd
import os

# Caminho para a pasta contendo os arquivos .nc
data_path = "D:/dados_nc"  # Altere para o caminho correto da sua pasta

# Listar os arquivos .nc na pasta
file_paths = [os.path.join(data_path, file) for file in os.listdir(data_path) if file.endswith(".nc")]

# Verificar se há arquivos .nc
if not file_paths:
    raise ValueError("Nenhum arquivo .nc encontrado na pasta.")

# Carregar cada arquivo individualmente e garantir que os índices de tempo sejam monotônicos
datasets = []
for file_path in file_paths:
    try:
        ds = xr.open_dataset(file_path, decode_times=False)
        
        # Garantir que a dimensão 'time' esteja monotônica
        if "time" in ds.variables:
            ds = ds.sortby("time")  # Ordenar por 'time'
        
        datasets.append(ds)
    except Exception as e:
        print(f"Falha ao carregar o arquivo {file_path}: {e}")

# Concatenar os datasets manualmente
try:
    combined_data = xr.concat(datasets, dim="time")  # Usar 'time' como dimensão de concatenação
    print("Arquivos combinados com sucesso!")
except Exception as e:
    raise ValueError(f"Falha ao combinar arquivos: {e}")

# Decodificar manualmente a variável 'time' usando cftime
if "time" in combined_data.variables:
    time_units = combined_data["time"].attrs.get("units", "")
    raw_time_values = combined_data["time"].values  # Valores brutos (números)

    # Identificar o tipo de unidade de tempo
    if "days since" in time_units:
        reference_date = pd.Timestamp(time_units.split("days since ")[1])  # Extrair a data de referência
        decoded_dates = [reference_date + pd.Timedelta(days=int(t)) for t in raw_time_values]
    elif "months since" in time_units:
        reference_date = pd.Timestamp(time_units.split("months since ")[1])  # Extrair a data de referência
        decoded_dates = [reference_date + pd.DateOffset(months=int(t)) for t in raw_time_values]
    else:
        raise ValueError(f"Unidades de tempo não suportadas: {time_units}")

    # Atualizar a variável 'time' no dataset
    combined_data["time"] = xr.Variable(
        dims=combined_data["time"].dims,
        data=decoded_dates,
        attrs=combined_data["time"].attrs
    )

# Converter para DataFrame
df = combined_data.to_dataframe()

# Salvar como CSV
output_path = os.path.join(data_path, "dados_combinados.csv")
df.to_csv(output_path)

print(f"Arquivos combinados e salvos como CSV em: {output_path}")
```

##4.2 Separação de arquivos CSV
Há alguns arquivos que após a conversão do formato NC para CSV, que ficaram com um tamanho muito grande, e então não puderam ser importados no GitHub, então foi criado esse script para separá-los em arquivos de 50 mb no máximo e enviar para o GitHub. 

```python
import pandas as pd
import os

def split_csv_by_size(input_file, output_dir, max_size_mb=50):
    """
    Separa um arquivo CSV grande em arquivos menores com base no tamanho máximo.

    :param input_file: Caminho para o arquivo CSV de entrada.
    :param output_dir: Diretório onde os arquivos divididos serão salvos.
    :param max_size_mb: Tamanho máximo de cada arquivo dividido (em MB).
    """
    # Converte o tamanho máximo para bytes
    max_size_bytes = max_size_mb * 1024 * 1024

    # Cria o diretório de saída, se ele não existir
    os.makedirs(output_dir, exist_ok=True)

    # Lê o cabeçalho do arquivo CSV
    with open(input_file, 'r') as f:
        header = f.readline()

    # Inicializa variáveis
    chunk_size = 10000  # Número inicial de linhas por chunk (ajustável)
    file_index = 1
    current_chunk = []
    current_size = len(header.encode('utf-8'))  # Começa com o tamanho do cabeçalho

    # Lê o arquivo CSV em chunks
    for chunk in pd.read_csv(input_file, chunksize=chunk_size):
        for _, row in chunk.iterrows():
            # Converte a linha para string e calcula seu tamanho em bytes
            row_str = ','.join(map(str, row.values)) + '\n'
            row_size = len(row_str.encode('utf-8'))

            # Verifica se a linha cabe no arquivo atual
            if current_size + row_size > max_size_bytes:
                # Salva o chunk atual como um novo arquivo
                output_file = os.path.join(output_dir, f'part_{file_index}.csv')
                with open(output_file, 'w') as f:
                    f.write(header)
                    f.writelines(current_chunk)
                
                # Reinicia variáveis para o próximo arquivo
                file_index += 1
                current_chunk = [row_str]
                current_size = len(header.encode('utf-8')) + row_size
            else:
                # Adiciona a linha ao chunk atual
                current_chunk.append(row_str)
                current_size += row_size

    # Salva o último chunk, se houver
    if current_chunk:
        output_file = os.path.join(output_dir, f'part_{file_index}.csv')
        with open(output_file, 'w') as f:
            f.write(header)
            f.writelines(current_chunk)

    print(f"Arquivo dividido em {file_index} partes.")

# Exemplo de uso
input_csv = "D:/dados_nc/.csv"
output_directory = "caminho/para/diretorio_saida"
split_csv_by_size(input_csv, output_directory, max_size_mb=50)
```

## 4.3 Busca de PIB e População por país
Script para buscar o PIB e a População por país e ano

```python
%sql
  CREATE TABLE IF NOT EXISTS workspace.pucrio.pib_populacao_anual(
    sigla_pais STRING,
    nome_pais STRING,
    ano INT,
    populacao BIGINT,
    pib_usd DOUBLE
  );
```

```python
import requests
import time
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, LongType, DoubleType

spark = SparkSession.builder.appName("PIB_Populacao_Historico").getOrCreate()

# === 1. Funções auxiliares ===

def fetch_country_list():
    """Busca lista de países ativos (exclui agregados e regiões)"""
    url = "https://api.worldbank.org/v2/country?format=json&per_page=300"
    try:
        r = requests.get(url, timeout=20)
        r.raise_for_status()
        data = r.json()
        countries = data[1]  # estrutura: [meta, lista]
        # Filtra países com ISO-3 válido e exclui agregados
        return [
            c for c in countries
            if c.get('id') and len(c['id']) == 3 and c['id'].isalpha() and c['id'] not in ['WLD', 'EUU', 'AFR', 'LCN', 'EAS']
        ]
    except Exception as e:
        raise Exception(f"Erro ao buscar lista de países: {e}")

def fetch_indicator(iso2, indicator_code, start_year=1950, end_year=2024):
    """Busca série temporal de um indicador para um país"""
    url = f"https://api.worldbank.org/v2/country/{iso2}/indicator/{indicator_code}?date={start_year}:{end_year}&format=json&per_page=100"
    data_points = {}
    page = 1
    max_pages = 5  # previne loop infinito

    while page <= max_pages:
        try:
            r = requests.get(f"{url}&page={page}", timeout=30)
            if r.status_code != 200:
                break
            data = r.json()
            if len(data) < 2 or not data[1]:
                break
            for d in data[1]:
                year = d.get('date')
                val = d.get('value')
                if year and val is not None:
                    try:
                        data_points[int(year)] = float(val)
                    except (ValueError, TypeError):
                        pass
            # Verifica paginação
            if 'pages' not in data[0] or page >= data[0]['pages']:
                break
            page += 1
            time.sleep(0.3)  # rate limiting leve
        except Exception as e:
            print(f"⚠️ Erro ao buscar {indicator_code} para {iso2}, página {page}: {e}")
            break
    return data_points

# === 2. Coleta dados ===

print("🔍 Obtendo lista de países...")
countries = fetch_country_list()
print(f"✅ Encontrados {len(countries)} países válidos.")

rows = []
total = len(countries)
for i, c in enumerate(countries, 1):
    iso2 = c.get('iso2Code')
    iso3 = c.get('id')
    name = c.get('name')
    
    if not iso2 or iso2 == 'NA' or not iso3 or len(iso3) != 3:
        continue
    
    print(f"[{i}/{total}] {name} ({iso3})...")
    
    # Busca população e PIB
    pop_data = fetch_indicator(iso2, 'SP.POP.TOTL')
    gdp_data = fetch_indicator(iso2, 'NY.GDP.MKTP.CD')
    
    # Anos comuns
    all_years = sorted(set(pop_data.keys()) | set(gdp_data.keys()))
    
    for year in all_years:
        rows.append({
            'sigla_pais': iso3.upper(),
            'nome_pais': name[:100],
            'ano': int(year),
            'populacao': int(pop_data.get(year)) if pop_data.get(year) else None,
            'pib_usd': float(gdp_data.get(year)) if gdp_data.get(year) else None
        })
    
    time.sleep(0.5)  # respeita rate limit da API (1 req/seg)

# === 3. Cria DataFrame Spark ===

schema = StructType([
    StructField("sigla_pais", StringType(), False),
    StructField("nome_pais", StringType(), False),
    StructField("ano", IntegerType(), False),
    StructField("populacao", LongType(), True),
    StructField("pib_usd", DoubleType(), True)
])

df = spark.createDataFrame(rows, schema)

# Remove duplicatas (ex: Z7 aparece mais de uma vez)
df = df.dropDuplicates(['sigla_pais', 'ano'])

print(f"✅ Coletados {df.count()} registros históricos (país × ano).")

# === 4. Salva como tabela Silver (ex: para junção com emissões) ===

df.write \
  .mode("overwrite") \
  .option("overwriteSchema", "true") \
  .saveAsTable("workspace.pucrio.pib_populacao_anual")

print("✅ Tabela `pib_populacao_anual` criada com sucesso.")
print("\n📌 Exemplo de uso:")
print("""SELECT * FROM workspace.pucrio.pib_populacao_anual 
WHERE sigla_pais = 'BRA' AND ano BETWEEN 2010 AND 2023;""")
```

##4.4. Ranking de PIB por países dos últimos 25 anos 
Esses scripts listam um ranking de PIB dos países dos últimos 25 anos pela tabela PIB_PAIS que foi importada em outro script.

```python
%sql
WITH ultimos_25_anos AS (
  SELECT 
    p.id_pais,
    a.ano,
    p.pib_pais
  FROM workspace.pucrio.PIB_PAIS p
  JOIN workspace.pucrio.ano a ON p.ID_ANO = a.id_ano
  WHERE a.ano >= 2000 AND a.ano <= 2024
    AND p.pib_pais IS NOT NULL
    AND p.pib_pais > 0
),
media_pib AS (
  SELECT 
    id_pais,
    AVG(pib_pais) AS pib_medio
  FROM ultimos_25_anos
  GROUP BY id_pais
  ORDER BY pib_medio DESC
  LIMIT 10
)
SELECT 
  u.id_pais,
  pa.nome_pais,
  u.ano,
  u.pib_pais
FROM ultimos_25_anos u
JOIN media_pib m ON u.id_pais = m.id_pais
JOIN workspace.pucrio.pais pa ON u.id_pais = pa.id_pais
ORDER BY m.pib_medio DESC, u.ano;
```

```python
import plotly.express as px
import pandas as pd

# 🔹 Executa a consulta SQL
df = spark.sql("""
WITH ultimos_25_anos AS (
  SELECT 
    p.id_pais,
    a.ano,
    p.pib_pais
  FROM workspace.pucrio.PIB_PAIS p
  JOIN workspace.pucrio.ano a ON p.ID_ANO = a.id_ano
  WHERE a.ano BETWEEN 2000 AND 2024
    AND p.pib_pais IS NOT NULL
    AND p.pib_pais > 0
),
media_pib AS (
  SELECT 
    id_pais,
    AVG(pib_pais) AS pib_medio
  FROM ultimos_25_anos
  GROUP BY id_pais
  ORDER BY pib_medio DESC
  LIMIT 10
)
SELECT 
  u.id_pais,
  pa.nome_pais,
  u.ano,
  u.pib_pais
FROM ultimos_25_anos u
JOIN media_pib m ON u.id_pais = m.id_pais
JOIN workspace.pucrio.pais pa ON u.id_pais = pa.id_pais
ORDER BY m.pib_medio DESC, u.ano
""").toPandas()

# 🔹 Verificação
if df.empty:
    raise ValueError("⚠️ Nenhum dado encontrado em PIB_PAIS para 2000–2024.")
print(f"✅ Top 10 países carregados | Anos: {df['ano'].min()}–{df['ano'].max()}")

# ✅ Gráfico interativo: linhas por país, eixo Y = PIB (USD)
fig = px.line(
    df,
    x='ano',
    y='pib_pais',
    color='nome_pais',
    title='<b>Top 10 Maiores Economias (PIB nominal em USD) — 2000–2024</b>',
    labels={
        'ano': 'Ano',
        'pib_pais': 'PIB (USD)',
        'nome_pais': 'País'
    },
    height=700,
    markers=True
)

# 🔧 Melhorias de layout
fig.update_layout(
    hovermode='x unified',
    legend_title='País',
    legend=dict(
        orientation='v',
        yanchor='top',
        y=1.0,
        xanchor='left',
        x=1.02,
        font=dict(size=10)
    ),
    xaxis=dict(dtick=2),
    yaxis=dict(
        title='PIB (USD)',
        tickformat='$,.0f',  # ex: $2,000,000,000,000
        rangemode='tozero'
    ),
    margin=dict(l=60, r=200, t=70, b=60)
)

# ✅ Exibir
fig.show()
```

#7. Catálogo de dados normalizados.
Aqui serão descritas as tabelas e campos que serão usados para normalizar os dados coletados nos Insights.

##7.1. Descrição tabelas

**REGIOES** - Representa regiões geográficas (continentes e sub-regiões) com base no catálogo de países.

**PAISES** - Dimensão de países, com dados demográficos e ligação à região.

**SETORES** - Setores de emissão (Power, Industry, etc.) — baseado em CARBON_MONITOR, CW, etc.

**FORMATO_VALORES** - Especifica unidades e fatores de conversão (ex: MtCO₂e → toneladas).

**PRODUTOS** -Produtos agrícolas e itens relacionados (ex: rice, meat_cattle), com ligação ao formato.

**ANOS** - Dimensão de tempo (ano), com validação de faixa.

**EMISSAO_CO2_ANUAL** - Emissões agregadas globais por ano (ex: WORLD de Carbon Monitor ou somatório OWID).

**EMISSAO_CO2_PER_CAPITA** - Valor médio global per capita por ano (ex: de OWID). 

**EMISSAO_CO2_SETOR** - Total por setor (global, ano agregado).

**EMISSAO_CO2_PRODUTOS** - Total por produto agrícola (ex: soma de emission_intensity_* × produção estimada ou intensidade média).

**EMISSAO_CO2_PAIS** - Emissões por país e ano (principal fato do modelo).

**EMISSAO_CO2_REGIAO** - Emissões por setor e ano (ex: Power em 2023).

**EMISSAO_CO2_ANUAL_PER_CAPITA** - Emissões por produto e ano (ex: rice em 1961, MtCO₂e).

**EMISSAO_CO2_ANUAL_PIB** - Emissões por país, ano e relacionadas ao PIB, para armazenar apenas o indicador per-PIB como fato independente.

**EMISSAO_CO2_ANUAL_SETOR** - Emissões por setor e ano (ex: Power em 2023).

**EMISSAO_CO2_ANUAL_PRODUTO** - Emissões por produto e ano (ex: rice em 1961, MtCO₂e).

**EMISSAO_CO2_ANUAL_PAIS** - Emissões por país e ano (principal fato do modelo).

**EMISSAO_CO2_ANUAL_REGIAO** - Emissões por região e ano (ex: Europa, 2020), obtida por soma agrupada de países.

**EMISSAO_CO2_PAIS_SETOR** - Emissões por país e setor (ano agregado ou média).

**EMISSAO_CO2_PAIS_PRODUTOS** - Emissões por país e produto agrícola (ex: meat_cattle no Brasil).

**EMISSAO_CO2_REGIAO_SETOR** - Emissões por região × setor (ex: Américas × Energy).

**EMISSAO_CO2_REGIAO_PRODUTOS** - Emissões por região × produto agrícola (ex: África × meat_cattle).





##7.2. SQL Criação tabelas

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.pucrio.regiao (
  id_regiao      BIGINT GENERATED ALWAYS AS IDENTITY,
  sigla_regiao   STRING NOT NULL,
  nome_regiao    STRING NOT NULL,
  NOME_REGIAO_INGLES STRING,
  CONSTRAINT pk_regiao PRIMARY KEY (id_regiao)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.pais (
  id_pais         BIGINT GENERATED ALWAYS AS IDENTITY,
  sigla_pais      STRING NOT NULL,          -- ISO Alpha-3 (ex: BRA)
  nome_pais       STRING NOT NULL,          -- Nome comum (ex: Brazil)
  id_regiao       BIGINT NOT NULL,
  CONSTRAINT pk_pais PRIMARY KEY (id_pais),
  CONSTRAINT fk_pais_regiao FOREIGN KEY (id_regiao) REFERENCES workspace.pucrio.regiao(id_regiao)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.ano (
  id_ano  BIGINT GENERATED ALWAYS AS IDENTITY,
  ano     INT NOT NULL,
  CONSTRAINT pk_ano PRIMARY KEY (id_ano)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.PIB_PAIS (
  ID_PIB_PAIS     BIGINT GENERATED ALWAYS AS IDENTITY,
  id_pais         BIGINT,
  ID_ANO          BIGINT,
  pib_pais        DOUBLE,                   -- PIB nominal (USD)
  CONSTRAINT pk_PIB_PAIS PRIMARY KEY (ID_PIB_PAIS),
  CONSTRAINT fk_PIB_PAIS_PAIS FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.PAIS(id_pais),
  CONSTRAINT fk_PIB_PAIS_ANO FOREIGN KEY (ID_ANO) REFERENCES workspace.pucrio.ANO(id_ano)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.POPULACAO_PAIS (
  ID_POPULACAO_PAIS BIGINT GENERATED ALWAYS AS IDENTITY,
  id_pais         BIGINT,
  ID_ANO          BIGINT,
  POPULACAO_PAIS DOUBLE,                   -- População
  CONSTRAINT pk_ID_POPULACAO_PAIS PRIMARY KEY (ID_POPULACAO_PAIS),
  CONSTRAINT fk_populacao_PAIS_PAIS FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.PAIS(id_pais),
  CONSTRAINT fk_populacao_PAIS_ANO FOREIGN KEY (ID_ANO) REFERENCES workspace.pucrio.ANO(id_ano)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.setor (
  id_setor        BIGINT GENERATED ALWAYS AS IDENTITY,
  sigla_setor     STRING NOT NULL,
  nome_setor      STRING NOT NULL,
  descricao_setor STRING,
  CONSTRAINT pk_setor PRIMARY KEY (id_setor)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.formato_valores (
  id_formato       BIGINT GENERATED ALWAYS AS IDENTITY,
  sigla_formato    STRING NOT NULL,      -- ex: MT, TON, KG
  nome_formato     STRING NOT NULL,      -- ex: 'Megatoneladas CO₂e'
  fator_conversao  DOUBLE NOT NULL,      -- para conversão → toneladas
  CONSTRAINT pk_formato PRIMARY KEY (id_formato)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.produto (
  id_produto            BIGINT GENERATED ALWAYS AS IDENTITY,
  sigla_produto         STRING NOT NULL,
  nome_produto          STRING NOT NULL,
  descricao_produto     STRING,
  id_formato_valores    BIGINT NOT NULL,
  CONSTRAINT pk_produto PRIMARY KEY (id_produto),
  CONSTRAINT fk_produto_formato FOREIGN KEY (id_formato_valores) REFERENCES workspace.pucrio.formato_valores(id_formato)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual (
  id_ano            BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE NOT NULL,  -- em toneladas (padrão)
  CONSTRAINT pk_anual PRIMARY KEY (id_ano),
  CONSTRAINT fk_anual_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES  workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_per_capita (
  id_ano            BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  ID_PAIS           BIGINT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_per_capita PRIMARY KEY (id_ano),
  CONSTRAINT fk_per_capita_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_per_capita_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE),
  CONSTRAINT FK_emissao_co2_per_capita_PAIS FOREIGN KEY (ID_PAIS) REFERENCES WORKSPACE.PUCRIO.PAIS (id_pais)
)
USING DELTA; 

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_setor (
  id_setor          BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_setor PRIMARY KEY (id_setor),
  CONSTRAINT fk_emissao_setor FOREIGN KEY (id_setor) REFERENCES workspace.pucrio.setor(id_setor),
  CONSTRAINT fk_emissao_setor_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_produto (
  id_produto        BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_produto PRIMARY KEY (id_produto),
  CONSTRAINT fk_emissao_produto FOREIGN KEY (id_produto) REFERENCES workspace.pucrio.produto(id_produto),
  CONSTRAINT fk_emissao_produto_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_pais (
  id_ano            BIGINT NOT NULL,
  id_pais           BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  valor_per_capita  DOUBLE,
  valor_per_pib     DOUBLE,
  CONSTRAINT pk_emissao_anual_pais PRIMARY KEY (id_ano, id_pais),
  CONSTRAINT fk_emissao_anual_pais_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_pais_pais FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.pais(id_pais),
  CONSTRAINT fk_emissao_anual_pais_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_setor (
  id_ano            BIGINT NOT NULL,
  id_setor          BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_anual_setor PRIMARY KEY (id_ano, id_setor),
  CONSTRAINT fk_emissao_anual_setor_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_setor_setor FOREIGN KEY (id_setor) REFERENCES workspace.pucrio.setor(id_setor),
  CONSTRAINT fk_emissao_anual_setor_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_produto (
  id_ano            BIGINT NOT NULL,
  id_produto        BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_anual_produto PRIMARY KEY (id_ano, id_produto),
  CONSTRAINT fk_emissao_anual_produto_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_produto_produto FOREIGN KEY (id_produto) REFERENCES workspace.pucrio.produto(id_produto),
  CONSTRAINT fk_emissao_anual_produto_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_pais_setor (
  id_pais           BIGINT NOT NULL,
  id_setor          BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_pais_setor PRIMARY KEY (id_pais, id_setor),
  CONSTRAINT fk_emissao_pais_setor_pais FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.pais(id_pais),
  CONSTRAINT fk_emissao_pais_setor_setor FOREIGN KEY (id_setor) REFERENCES workspace.pucrio.setor(id_setor),
  CONSTRAINT fk_emissao_pais_setor_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_pais_produto (
  id_pais           BIGINT NOT NULL,
  id_produto        BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,
  CONSTRAINT pk_emissao_pais_produto PRIMARY KEY (id_pais, id_produto),
  CONSTRAINT fk_emissao_pais_produto_pais FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.pais(id_pais),
  CONSTRAINT fk_emissao_pais_produto_produto FOREIGN KEY (id_produto) REFERENCES workspace.pucrio.produto(id_produto),
  CONSTRAINT fk_emissao_pais_produto_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_regiao_setor (
  id_regiao     BIGINT NOT NULL,
  id_setor      BIGINT NOT NULL,
  id_fonte_dados INT NOT NULL,
  valor_emissao DOUBLE,  -- em toneladas (CO₂ ou CO₂e, conforme fonte)
  CONSTRAINT pk_emissao_regiao_setor PRIMARY KEY (id_regiao, id_setor),
  CONSTRAINT fk_emissao_regiao_setor_regiao FOREIGN KEY (id_regiao) REFERENCES workspace.pucrio.regiao(id_regiao),
  CONSTRAINT fk_emissao_regiao_setor_setor FOREIGN KEY (id_setor) REFERENCES workspace.pucrio.setor(id_setor),
  CONSTRAINT fk_emissao_regiao_setor_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_regiao_produto (
  id_regiao     BIGINT NOT NULL,
  id_produto    BIGINT NOT NULL,
  id_fonte_dados INT NOT NULL,
  valor_emissao DOUBLE,  -- em toneladas (CO₂e)
  CONSTRAINT pk_emissao_regiao_produto PRIMARY KEY (id_regiao, id_produto),
  CONSTRAINT fk_emissao_regiao_produto_regiao FOREIGN KEY (id_regiao) REFERENCES workspace.pucrio.regiao(id_regiao),
  CONSTRAINT fk_emissao_regiao_produto_produto FOREIGN KEY (id_produto) REFERENCES workspace.pucrio.produto(id_produto),
  CONSTRAINT fk_emissao_regiao_produto_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_pais (
  id_ano            BIGINT NOT NULL,
  id_pais           BIGINT NOT NULL,
  id_fonte_dados    INT NOT NULL,
  valor_emissao     DOUBLE,  -- CO₂ total (excl. LULUCF), em toneladas
  valor_per_capita  DOUBLE,  -- toneladas / pessoa
  valor_per_pib     DOUBLE,  -- toneladas / mil USD (PIB nominal)
  CONSTRAINT pk_emissao_anual_pais PRIMARY KEY (id_ano, id_pais),
  CONSTRAINT fk_emissao_anual_pais_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_pais_pais FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.pais(id_pais),
  CONSTRAINT fk_emissao_anual_pais_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_regiao (
  id_ano        BIGINT NOT NULL,
  id_regiao     BIGINT NOT NULL,
  id_fonte_dados INT NOT NULL,
  valor_emissao DOUBLE,  -- em toneladas
  CONSTRAINT pk_emissao_anual_regiao PRIMARY KEY (id_ano, id_regiao),
  CONSTRAINT fk_emissao_anual_regiao_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_regiao_regiao FOREIGN KEY (id_regiao) REFERENCES workspace.pucrio.regiao(id_regiao),
  CONSTRAINT fk_emissao_anual_regiao_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;

CREATE TABLE IF NOT EXISTS workspace.pucrio.emissao_co2_anual_pib (
  id_ano        BIGINT NOT NULL,
  id_pais       BIGINT NOT NULL,
  id_fonte_dados INT NOT NULL,
  valor_emissao DOUBLE,  -- valor_per_pib já em toneladas / mil USD
  CONSTRAINT pk_emissao_fato_anual_pib PRIMARY KEY (id_ano, id_pais),
  CONSTRAINT fk_emissao_anual_pib_ano FOREIGN KEY (id_ano) REFERENCES workspace.pucrio.ano(id_ano),
  CONSTRAINT fk_emissao_anual_pib_pais FOREIGN KEY (id_pais) REFERENCES workspace.pucrio.pais(id_pais),
  CONSTRAINT fk_emissao_anual_pib_fonte_dados FOREIGN KEY (id_fonte_dados) REFERENCES workspace.pucrio.catalogo_fontes_dados(ID_FONTE)
)
USING DELTA;
```

#8. Scripts SQL´s para Insights 
Foram criados alguns SQL´s para os insigths de dados,

**- Estão apenas os SQL´s relacionaddos à dados que contenham informações de emissão de dióxido de carbono (CO2)**
**- Caso os dados e informações coletadas sejam de anos anteriores à 1950, os mesmos serão filtrados/ignorados.**

Cada consulta irá realizar uma inserção dos dados em tabelas criadas e normalizadas para a armazenagem correta dos dados

##Detalhamento dos campos e valores
Abaixo estará um valor com os detalhamentos de valores, campos, e formatos de gravações nas tabelas, para fins de padronização.


###Para os valores de MtCO2/day (MtCO2/dia): 
MtCO2/day significa megatoneladas de dióxido de carbono por dia. É uma unidade de medida utilizada para quantificar a taxa de emissões de dióxido de carbono (CO₂) em larga escala, geralmente em contextos de monitoramento global, nacional ou industrial das mudanças climáticas. 
A notação se divide da seguinte forma:
* Mt significa megatoneladas, ou seja, um milhão de toneladas métricas (1.000.000 de toneladas).
* CO₂ refere-se ao dióxido de carbono, o principal gás de efeito estufa emitido por atividades humanas, como a queima de combustíveis fósseis e a produção de cimento.
* /day indica a medição por dia (taxa diária). 


###Para os valores de CO2eq/kg
O termo CO2eq/kg (ou CO₂e/kg) significa "Dióxido de Carbono Equivalente por quilograma" e é uma métrica usada para medir e comparar o impacto ambiental de diferentes gases de efeito estufa (GEE) em uma única unidade comum. 

Funciona da seguinte forma:
* O que é CO₂e? Nem todos os gases de efeito estufa (como metano, óxido nitroso e outros) retêm calor na atmosfera com a mesma intensidade ou pelo mesmo período de tempo que o dióxido de carbono (CO2). 
* O CO₂e converte o potencial de aquecimento global (GWP) de todos esses outros gases em uma quantidade equivalente de (CO2) que produziria o mesmo efeito.O que é /kg? A divisão por quilograma indica a intensidade dessa emissão por unidade de peso do produto ou material em análise.
* Por exemplo, ela permite dizer que a produção de 1 kg de carne bovina gera uma quantidade de gases de efeito estufa equivalente a 1 kg de CO₂, que é muito maior do que a emissão gerada pela produção de 1 kg de legumes.

Exemplo Prático Suponha que você tenha: Intensidade de emissão para produzir aço: 1,85kg COeq/kg
Massa total de aço produzida pela fábrica: 500.000 kg

Exemplo do cálculo: 
Emissão total em kg: 
1,85 x 500.000 = 925.000 kg de COeq

Conversão para toneladas:
925.000 kg / 1000 = 925 toneladas de COeq

Com isso, a emissão total da fábrica foi de 925 toneladas de CO₂e.


###Para os valores gravados nas tabelas
* A padronização dos valores gravados nas tabelas serão de toneladas, então todos os valores dos dados coletados serão convertidos para toneladas.
* Um estudo sugere que (1) tonelada de (CO2) pode ser compensada por (7,14) árvores da Mata Atlântica
* Cada tonelada de (CO2) equivale a (1) crédito de carbono, que pode ser negociado no mercado.
* São necessários cerca de 435 litros de gasolina para produzir 1 tonelada (1.000kg) de CO2.

Histórico de emissões de dióxido de carbono diário (**MtCO2/day**) por países e setores, entretanto na consulta abaixo o consumo foi agregado em uma soma para contabilizar anualmente os valores e com isso serão realizados as inserções nas seguintes tabelas:

**Os valores serão convertidos em toneladas**

```python
%sql
--CONSULTA GERAL DO CARBON MONITOR COM TODOS OS DADOS E TODAS AS COLUNAS
SELECT CASE WHEN COUNTRY = 'EU27 & UK' THEN 'EUR' ELSE COUNTRY END PAIS, 
       CAST(substr(DATE, 7, 10) AS NUMERIC) ANO,
       SECTOR SETOR,
       SUM(VALUE) VALOR --CO2 emitidos por ano MtCO2/day
FROM workspace.pucrio.CARBON_MONITOR
GROUP BY 
       COUNTRY, 
       CAST(substr(DATE, 7, 10) AS NUMERIC),
       SECTOR
       ; 
```

Insight de Emissão de CO2 anual já convertidos em toneladas 

```python
%sql
--COM BASE NA CONSULTA ANTERIOR, ESTA SERÁ A INSERÇÃO NO EMISSAO_CO2_ANUAL
INSERT INTO WORKSPACE.PUCRIO.EMISSAO_CO2_ANUAL (ID_ANO, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       1,
       SUM(carbon_monitor.value) * formato_valores.fator_conversao VALOR_CONVERTIDO --CO2 emitidos por ano de CO2 em Toneladas
  FROM workspace.pucrio.CARBON_MONITOR
 INNER JOIN workspace.pucrio.ANO 
    ON ANO.ANO = CAST(substr(carbon_monitor.DATE, 7, 10) AS NUMERIC) 
 INNER JOIN workspace.pucrio.formato_valores ON ID_FORMATO = 1
 GROUP BY 
       ANO.ID_ANO,
       formato_valores.fator_conversao
       ; 
```

Insight na tabela de inserção de emissão de CO2 por setor que emite dióxido de carbono, com o valor convertido em Toneladas

```python
%sql
--COM BASE NA CONSULTA ANTERIOR, ESTA SERÁ A INSERÇÃO NA TABELA DE emissao_co2_setor 
INSERT INTO workspace.pucrio.emissao_co2_setor (id_setor, id_fonte_dados, valor_emissao) 
SELECT SETOR.ID_SETOR,    
       1,   
       SUM(carbon_monitor.VALUE) * formato_valores.fator_conversao VALOR_CONVERTIDO 
  FROM workspace.pucrio.CARBON_MONITOR 
 INNER JOIN workspace.pucrio.formato_valores 
    ON formato_valores.ID_FORMATO = 1
 INNER JOIN workspace.pucrio.SETOR
    ON UPPER(SETOR.nome_setor) = UPPER(carbon_monitor.sector)
GROUP BY  
      SETOR.id_setor,
      formato_valores.fator_conversao
      ; 
```

Insight na tabela de inserção de emissão de CO2 anual e por PIB dos países, o valor convertido será em valor de toneladas por milhão de Dólares no respectivo ano e país.

* Leva-se em consideração o preço do carbono em 30 dólares por tonelada. (valor referência do Brasil em 2025)


```python
%sql
--COM BASE NA CONSULTA ANTERIOR, ESTA SERÁ A INSERÇÃO NA TABELA DE emissao_co2_anual_pib 
INSERT INTO workspace.pucrio.emissao_co2_anual_pib (id_ano, ID_PAIS, id_fonte_dados, valor_emissao) 
SELECT ANO.ID_ANO,    
       PAIS.ID_PAIS,
       1,   
       (((SUM(carbon_monitor.VALUE) * formato_valores.fator_conversao)) * 1e6) / PAIS.pib_pais valor_t_por_milhao_USD 
  FROM workspace.pucrio.CARBON_MONITOR 
 INNER JOIN workspace.pucrio.formato_valores 
    ON formato_valores.ID_FORMATO = 1
 INNER JOIN workspace.pucrio.ANO 
    ON ANO.ANO = CAST(substr(carbon_monitor.DATE, 7, 10) AS NUMERIC)  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(CASE WHEN carbon_monitor.COUNTRY = 'EU27 & UK' THEN 'Europe' ELSE CARBON_MONITOR.COUNTRY END) = UPPER(PAIS.nome_pais)
 INNER JOIN workspace.pucrio.pib_pais
    ON PAIS.ID_PAIS = pib_pais.ID_PAIS
   AND ANO.ID_ANO = pib_pais.ID_ANO 
GROUP BY  
      ANO.ID_ANO,  
      formato_valores.fator_conversao,  
      PAIS.ID_PAIS,
      PAIS.pib_pais    ; 
```

Insight na tabela de inserção de emissão de CO2 anual e por região, com o valor convertido em Toneladas

```python
%sql
--COM BASE NA CONSULTA ANTERIOR, ESTA SERÁ A INSERÇÃO NA TABELA DE emissao_co2_anual_regiao 
INSERT INTO workspace.pucrio.emissao_co2_anual_regiao (id_ano, id_regiao, id_fonte_dados, valor_emissao) 
SELECT ANO.ID_ANO,    
       REGIAO.ID_REGIAO,
       1,   
       ((SUM(carbon_monitor.VALUE) * formato_valores.fator_conversao)) VALOR_CONVERTIDO 
  FROM workspace.pucrio.CARBON_MONITOR 
 INNER JOIN workspace.pucrio.formato_valores 
    ON formato_valores.ID_FORMATO = 1
 INNER JOIN workspace.pucrio.ANO 
    ON ANO.ANO = CAST(substr(carbon_monitor.DATE, 7, 10) AS NUMERIC)  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(CASE WHEN carbon_monitor.COUNTRY = 'EU27 & UK' THEN 'Europe' ELSE CARBON_MONITOR.COUNTRY END) = UPPER(PAIS.nome_pais)
 INNER JOIN workspace.pucrio.regiao
    ON PAIS.ID_REGIAO = REGIAO.ID_REGIAO     
GROUP BY  
      ANO.ID_ANO,  
      formato_valores.fator_conversao,  
       REGIAO.ID_REGIAO 
      ; 
```

Insight na tabela de inserção de emissão de CO2 per capita (população), com o valor convertido em Toneladas per capita

```python
%sql
--COM BASE NA CONSULTA ANTERIOR, ESTA SERÁ A INSERÇÃO NA TABELA DE emissao_co2_per_capita 
INSERT INTO workspace.pucrio.emissao_co2_per_capita (id_ano, ID_PAIS, id_fonte_dados, valor_emissao) 
SELECT ANO.ID_ANO,    
       PAIS.ID_PAIS,
       1,    
      ((SUM(carbon_monitor.VALUE) * formato_valores.fator_conversao)) / populacao_pais.POPULACAO_PAIS AS valor_t_por_habitante 
  FROM workspace.pucrio.CARBON_MONITOR 
 INNER JOIN workspace.pucrio.formato_valores        
    ON formato_valores.ID_FORMATO = 1
 INNER JOIN workspace.pucrio.ANO 
    ON ANO.ANO = CAST(substr(carbon_monitor.DATE, 7, 10) AS NUMERIC)  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(CASE WHEN carbon_monitor.COUNTRY = 'EU27 & UK' THEN 'Europe' ELSE CARBON_MONITOR.COUNTRY END) = UPPER(PAIS.nome_pais)
 INNER JOIN workspace.pucrio.populacao_pais
    ON PAIS.ID_PAIS = populacao_pais.ID_PAIS
   AND ANO.id_ano = populacao_pais.id_ano
 GROUP BY  
      ANO.ID_ANO,    
      PAIS.ID_PAIS,
      formato_valores.fator_conversao,
      populacao_pais.POPULACAO_PAIS 
      ; 
```

###8.3.2. HISTORICAL EMISSIONS

Histórico de emissões de CO₂e por quilograma por setor, valor de referência é: CO2eq/kg por países e setores que emitem CO2, entretanto na consulta abaixo o consumo foi agregado em uma soma para contabilizar anualmente os valores em toneladas e com isso serão realizados as inserções nas seguintes tabelas:

* emissao_co2_anual
* emissao_co2_anual_pais
* emissao_co2_anual_pib
* emissao_co2_anual_setor
* emissao_co2_anual_regiao
* emissao_co2_pais_setor
* emissao_co2_per_capita
* emissao_co2_setor
* emissao_co2_regiao_setor



Consulta geral do histórico de emissões de carbono por setor, por ano e filtrado apenas o gás de CO2.

* O valor de referência está em MtCO₂e 
* Os valores serão converrtidos em toneladas
* ID da fonte de dados: 10 - CW_HISTORICAL_EMISSIONS

```python
%sql
WITH EMISSIONS AS (       
       SELECT ANO, 
              GAS,
              SECTOR,
              SIGLA_PAIS,
              SUM(CAST(VALOR AS NUMERIC)) VALOR
       FROM (
       SELECT YEAR ANO,  
              GAS,
              SECTOR,
              ISO SIGLA_PAIS, 
              EMISSIONS VALOR --CO2  MtCO₂e
              FROM workspace.pucrio.cw_historical_emissions   
              UNPIVOT (
              EMISSIONS FOR YEAR IN (
                     `1961`, `1962`, `1963`, `1964`, `1965`, `1966`, `1967`, `1968`, `1969`, `1970`,
                     `1971`, `1972`, `1973`, `1974`, `1975`, `1976`, `1977`, `1978`, `1979`, `1980`,
                     `1981`, `1982`, `1983`, `1984`, `1985`, `1986`, `1987`, `1988`, `1989`, `1990`,
                     `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`,
                     `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`,
                     `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`,
                     `2021`, `2022`, `2023`
              )
              ) AS unpivoted_data  
       )   
       WHERE UPPER(GAS) LIKE '%CO2%' 
       AND UPPER(VALOR) NOT IN ('N/A') 
       AND UPPER(VALOR) IS NOT NULL
       GROUP BY ANO, 
              GAS,
              SECTOR,
              SIGLA_PAIS  
)
SELECT B.*
  FROM EMISSIONS B
WHERE B.sigla_pais NOT IN ('BASIC', 'NONANNEXI', 'WORLD', 'ANNEXI', 'AOSIS', 'UMBRELLA');
```

* emissao_co2_produto
* emissao_co2_pib
* emissao_co2_per_capita
* emissao_co2_pais_produto

Observações e detalhes 
* Os valores deste arquivo de dados estão no formato: kt = kilotonelada métrica, ou seja, 1.000 toneladas métricas — padrão usado pela FAO, ONU, IEA, etc.
* Exemplo de uso: Se um país emitiu 125,7 kt CO₂, então: 125,7 × 1,000 = 125,700 toneladas de CO₂ 
* ID da fonte de dados: 13 - FAOSTAT_EMISSIONS_TOTALS

Abaixo o SQL geral de emissaão

```python
%sql
WITH EMISSIONS AS (
       SELECT 
              AREA PAIS, 
              substr(YEAR, 2, 4) ANO, 
              ITEM PRODUTO,
              ELEMENT GAS,
              EMISSIONS VALOR --CO2  KT (kt)
       FROM faostat_emissions_totals
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `Y1961`, `Y1962`, `Y1963`, `Y1964`, `Y1965`, `Y1966`, `Y1967`, `Y1968`, `Y1969`, `Y1970`,
              `Y1971`, `Y1972`, `Y1973`, `Y1974`, `Y1975`, `Y1976`, `Y1977`, `Y1978`, `Y1979`, `Y1980`,
              `Y1981`, `Y1982`, `Y1983`, `Y1984`, `Y1985`, `Y1986`, `Y1987`, `Y1988`, `Y1989`, `Y1990`,
              `Y1991`, `Y1992`, `Y1993`, `Y1994`, `Y1995`, `Y1996`, `Y1997`, `Y1998`, `Y1999`, `Y2000`,
              `Y2001`, `Y2002`, `Y2003`, `Y2004`, `Y2005`, `Y2006`, `Y2007`, `Y2008`, `Y2009`, `Y2010`,
              `Y2011`, `Y2012`, `Y2013`, `Y2014`, `Y2015`, `Y2016`, `Y2017`, `Y2018`, `Y2019`, `Y2020`,
              `Y2021`, `Y2022`, `Y2023`, `Y2030`, `Y2050` 
       )
       ) AS unpivoted_data
       WHERE `Element Code` IN ('724313','723113','724413','7273','717815')
)
SELECT B.*
  FROM EMISSIONS B;
```

Os dados quase serão ignorados, pois não mostram a quantidade de emissão de CO2 por países, renda per capita, PIB e/ou outro indicador, apenas por área e latitude e logitude e a área, e isso acabou inviabilizando o projeto, pois mesmo com um script em python bem refinado ainda poderá haver situações em que as informações ficarão erradas para certos países, e isso acabará que mostrando valores inconsistentes.

Entretanto, o valor destas informações obtidas na NASA são muito interessantes, e com isso poderá ser usado um consumo anual/mensal/diário para o planeta todo

Também foi identificado valores de emissão de Dióxido de Carbono (CO2) negativos, estes valores negativos são provinientes de latitudes e longitudes em locais como: Antartida (extremo polo sul), Groelândia (extremo polo norte), e alguns locais de alto mar, até mesmo próximos à linha do Equador. Isso mostra a importância destes dados, pois com estas informações e mais algumas pesquisas científicas (não tão aprofundadas), passamos a entender a importância do derretimento das calotas polares, aquecimento dos mares e o aumento do nível do mar.

**Conclusão**
Quanto à este dado da Nasa, quanto menor o valor (mais negativo for) para estes locais, melhor é para a saúde geral do planeta (no caso para o aquecimento global), pois estará absorvendo mais raios ultravioletas e mantendo a temperatura geral do planeta estável. 

**Observação**
A consulta e a inclusão das informações abaixo estão em um nível agregado de valor anual, ou seja, para conseguir chegar no valor de emissão de CO2 negativo, precisa retirar a agregação.

```python
%sql
SELECT ANO,
       SUM(VALOR) VALOR
FROM (
       SELECT CAST(substr(CAST(TRUNC(TIME, 'YEAR') AS STRING), 1, 4) AS NUMERIC) ANO,
              total_post VALOR, --g/m^2/day
              area qtd_area_m2
       FROM workspace.pucrio.nasa_cms_flux_total
)
GROUP BY ANO
ORDER BY 1;
```

##8.6. ONU

Dados das nações unidas são muito valiosos, pois a instituição/orgaização fornece várias informações gratuitamente e é mantido por vários governos no mundo inteiro.

* As informações de valores são em kt - Kilotoneladas de CO2 

**Explicação detalhada dos dados coletados**

CO2 total with LULUCF" refere-se ao cálculo das emissões e remoções líquidas totais de dióxido de carbono (CO2) de um país ou região, incluindo as atividades do setor de Uso da Terra, Mudança do Uso da Terra e Florestas (do inglês, Land Use, Land-Use Change and Forestry - LULUCF). 

A sigla LULUCF abrange todas as atividades relacionadas ao uso e manejo do solo, que podem atuar tanto como fontes de emissão quanto como sumidouros de CO2. 

Fontes de emissão: Atividades como desmatamento, queima de biomassa, ou a drenagem de áreas úmidas (como turfeiras) liberam CO2 na atmosfera.
Sumidouros de carbono: Processos naturais, como o crescimento de florestas e a absorção de carbono pelo solo, removem o CO2 da atmosfera. 

A inclusão do LULUCF no cálculo total do CO2 é crucial porque fornece uma imagem mais completa do balanço de carbono de uma área:
CO2 total (excluindo LULUCF): Considera apenas as emissões provenientes da queima de combustíveis fósseis, processos industriais e outras fontes não biológicas.

CO2 total (com LULUCF): Inclui o balanço líquido do setor LULUCF (emissões menos remoções). Isso significa que, se um país tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, o valor líquido do LULUCF pode ser negativo (uma remoção), ajudando a compensar as emissões de outros setores. 

###8.6.1. HISTORICAL EMISSIONS

Histórico de emissões de dióxido de carbono em CO2 Kt, por países e setores, entretanto na consulta abaixo o consumo foi agregado em uma soma para contabilizar anualmente os valores e com isso serão realizados as inserções nas seguintes tabelas:

* emissao_co2_anual
* emissao_co2_anual_pais
* emissao_co2_anual_pib
* emissao_co2_anual_regiao
* emissao_co2_per_capita

Abaixo a consulta geral

* Fonte de dados: 17 - ONU_TIME_SERIES_CO_2_TOTAL 

```python
%sql
WITH EMISSIONS AS (
       SELECT 
              PARTY PAIS,
              YEAR ANO,  
              EMISSIONS VALOR, --KT
              `Base Year` VALOR_ANO_BASE, --kt
              `Change from base year to latest reported year` COMPARATIVO_ANO_BASE_ULTIMO_ANO, --%
              `Last Inventory Year 2021` ANO_ULTIMO_INVENTARIO --%      
       FROM onu_time_series_co_2_total
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1990`, `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`, `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`, `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`
       )
       ) AS unpivoted_data
)
SELECT B.*
  FROM EMISSIONS B;
```

Histórico de emissões de dióxido de carbono em CO2 Kt agregado anualmente (Tabela: emissao_co2_anual)

```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_anual (id_ano, id_fonte_dados, valor_emissao)
WITH EMISSIONS AS (
    SELECT PAIS, 
           ANO,
           CAST(replace(replace(VALOR, '.', ''), ',', '.') AS NUMERIC(20,5)) VALOR
     FROM (
       SELECT 
              PARTY PAIS,
              YEAR ANO,  
              EMISSIONS VALOR, --KT
              `Base Year` VALOR_ANO_BASE, --kt
              `Change from base year to latest reported year` COMPARATIVO_ANO_BASE_ULTIMO_ANO, --%
              `Last Inventory Year 2021` ANO_ULTIMO_INVENTARIO --%      
       FROM workspace.pucrio.onu_time_series_co_2_total
       UNPIVOT (
       EMISSIONS FOR YEAR IN (
              `1990`, `1991`, `1992`, `1993`, `1994`, `1995`, `1996`, `1997`, `1998`, `1999`, `2000`, `2001`, `2002`, `2003`, `2004`, `2005`, `2006`, `2007`, `2008`, `2009`, `2010`, `2011`, `2012`, `2013`, `2014`, `2015`, `2016`, `2017`, `2018`, `2019`, `2020`
       )
       ) AS unpivoted_data
     )
)
SELECT ANO.id_ano, 
       17, --FONTE DE DADOS
       SUM(B.VALOR * FORMATO_VALORES.fator_conversao) VALOR_CONVERTIDO
  FROM EMISSIONS B
 INNER JOIN workspace.PUCRIO.FORMATO_VALORES
    ON FORMATO_VALORES.ID_FORMATO = 2
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = B.ANO   
 GROUP BY ANO.id_ano  
 HAVING 
       SUM(B.VALOR * FORMATO_VALORES.fator_conversao)  > 0
```

Histórico de emissões de dióxido de carbono em CO2 Kt agregado anualmente e por país (Tabela: emissao_co2_anual_pais)

###8.7.2. CO2 EMISSIONS SHARE ANNUAL CO2 (dados ignorados)

A participação das emissões globais de CO₂ refere-se à percentagem ou quantidade de dióxido de carbono que um país, setor (como energia, transporte, indústria) ou atividade humana contribui para o total das emissões mundiais, impactando o efeito estufa e o aquecimento global, sendo o uso de combustíveis fósseis (carvão, petróleo, gás) a principal fonte. Isso ajuda a entender quem são os maiores poluidores e onde focar os esforços de redução, como no Brasil, que tem um papel importante por suas florestas e energias renováveis. 

O que Significa:
* Emissões de CO₂: A liberação de dióxido de carbono (CO₂) na atmosfera, principalmente pela queima de combustíveis fósseis, desmatamento e processos industriais.
* Participação Global: A fatia que cada um (país, setor) representa no total de CO₂ emitido por todas as fontes no mundo.
* Gás de Efeito Estufa (GEE): O CO₂ é o principal GEE, responsável por cerca de 74% do aquecimento global, prendendo calor na atmosfera e causando mudanças climáticas. 

Observações e detalhes:
* Valores estão em %
* A fonte de dados: 24 - OWD_SHARE_OF_CUMULATIVE_CO_2

```python
%sql
SELECT CODE PAIS, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS
       ENTITY NOME_PAIS, 
       YEAR ANO,       
       SUM(`Share of global annual CO₂ emissions`) VALOR
  FROM workspace.pucrio.owd_annual_share_of_co_2_emissions
 GROUP BY CODE,
       YEAR,
       Entity;     
```

###8.7.4. CO2 CONCENTRATION LONG TERM (ONLY WORLD) - (dados ignorados)

A concentração de dióxido de carbono (CO₂) na atmosfera é medida em partes por milhão (ppm). Tendências de longo prazo nas concentrações de CO₂ podem ser medidas com alta resolução usando amostras de ar preservadas em núcleos de gelo.

Baseado em estudos de núcleos de gelo sobre a concentração histórica de gases de efeito estufa e em locais de monitoramento do ar recentes ao redor do mundo.

Como esses dados são descritos pelo produtor?

Este indicador descreve como os níveis dos principais gases de efeito estufa (GEE) na atmosfera mudaram ao longo do tempo geológico e nos últimos anos. Mudanças nos GEE atmosféricos, em parte causadas por atividades humanas, afetam a quantidade de energia armazenada no sistema Terra-atmosfera e, portanto, afetam o clima da Terra. Este indicador é altamente relevante para as mudanças climáticas, pois os gases de efeito estufa provenientes de atividades humanas são o principal fator das mudanças climáticas observadas desde meados do século XX (IPCC, 2021).

**Conclusão**
Os dados estão incompletos e possuí apneas dados do mundo todo e a conversão dos dados são mensurados por partes por milhão de concetração de CO2, que é o mesmo que é lançado na atmosfera.

```python
%sql
SELECT YEAR ANO, --HÁ ANOS NEGATIVOS E ANOS COM VALORES MUITO BAIXOS
       SUM(`Annual concentration of atmospheric carbon dioxide`) VALOR
  FROM workspace.pucrio.owd_co_2_long_term_concentration
 WHERE YEAR > 1950
 GROUP BY YEAR;     
```

###8.7.5. CO2 EMISSIONS PER CAPITA

Em que lugar do mundo a pessoa média emite mais dióxido de carbono (CO2) por ano?

Podemos calcular a contribuição do cidadão médio de cada país dividindo suas emissões totais pela sua população. Isso nos dá as emissões de CO2 per capita. Na visualização, vemos as diferenças nas emissões per capita em todo o mundo.

Aqui, analisamos as emissões baseadas na produção – ou seja, as emissões produzidas dentro das fronteiras de um país, sem levar em conta como as mercadorias são comercializadas internacionalmente. Em nossa publicação sobre emissões baseadas no consumo, analisamos como esses números mudam quando consideramos o comércio. Os números da produção são importantes – são esses os números que são levados em conta para as metas climáticas¹ e, graças às reconstruções históricas, estão disponíveis para o mundo todo desde meados do século XVIII.

**Tabelas usadas: **
* emissao_co2_per_capita

**Observações e detalhes dos dados: **
* Os valores dos dados estão em toneladas
* Fonte de dados: 22 - OWD_CO_EMISSIONS_PER_CAPITA


```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_per_capita (id_ano, ID_PAIS, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,
       pais.id_pais, --SE ESTIVER NULO É UMA REGIÃO OU CONTINENTE E ENTÃO USAR O NOME DO PAÍS 
       22,
       SUM(`Annual CO₂ emissions (per capita)`) VALOR
  FROM  workspace.pucrio.owd_co_emissions_per_capita 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_co_emissions_per_capita.YEAR 
 INNER JOIN workspace.pucrio.pais
    ON UPPER(PAIS.sigla_pais) = UPPER(owd_co_emissions_per_capita.Code)
 WHERE CODE IS NOT NULL 
   AND owd_co_emissions_per_capita.Year > 1950   
 GROUP BY 
        ANO.ID_ANO,
        pais.id_pais
 HAVING SUM(`Annual CO₂ emissions (per capita)`) > 0;  
```

```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_anual(id_ano, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       23,
       SUM(`Cumulative CO₂ emissions`) VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_cumulative_co_emissions.YEAR  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(PAIS.sigla_pais) = UPPER(owd_cumulative_co_emissions.Code)
 WHERE CODE IS NOT NULL 
   AND owd_cumulative_co_emissions.Year > 1950   
 GROUP BY 
        ANO.ID_ANO  
 HAVING 
       SUM(`Cumulative CO₂ emissions`) > 0;  
```

Emissão de CO2 produzidas por combústiveis fósseis anual por país (Tabela: emissao_co2_anual_pais)


```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_anual_pais(id_ano, id_pais, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       PAIS.ID_PAIS,
       23,
       SUM(`Cumulative CO₂ emissions`) VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_cumulative_co_emissions.YEAR  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(PAIS.sigla_pais) = UPPER(owd_cumulative_co_emissions.Code)
 WHERE CODE IS NOT NULL 
   AND owd_cumulative_co_emissions.Year > 1950   
 GROUP BY 
       PAIS.ID_PAIS,
        ANO.ID_ANO  
 HAVING 
       SUM(`Cumulative CO₂ emissions`) > 0;  
```

Emissão de CO2 produzidas por combústiveis fósseis anual por país (Tabela: emissao_co2_anual_pib)



```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_anual_pib(id_ano, id_pais, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       PAIS.ID_PAIS,
       23,
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) / pib_pais.pib_pais VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_cumulative_co_emissions.YEAR  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(PAIS.sigla_pais) = UPPER(owd_cumulative_co_emissions.Code)
 INNER JOIN workspace.pucrio.pib_pais
    ON pib_pais.ID_PIB_PAIS = pais.id_pais
   AND PIB_PAIS.ID_ANO = ano.id_ano 
 WHERE CODE IS NOT NULL 
   AND owd_cumulative_co_emissions.Year > 1950   
 GROUP BY 
       PAIS.ID_PAIS,
        ANO.ID_ANO,
        pib_pais.pib_pais 
 HAVING 
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) / pib_pais.pib_pais > 0;  
```

Emissão de CO2 produzidas por combústiveis fósseis anual por país (Tabela: emissao_co2_anual_regiao)




```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_anual_regiao(id_ano, id_regiao, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       REGIAO.ID_REGIAO,
       23,
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_cumulative_co_emissions.YEAR   
 INNER JOIN workspace.pucrio.regiao
    ON UPPER(REGIAO.NOME_REGIAO_INGLES) LIKE '%' || UPPER(owd_cumulative_co_emissions.Entity) ||'%'
 WHERE CODE IS NULL 
   AND owd_cumulative_co_emissions.Year > 1950   
 GROUP BY 
       ANO.ID_ANO,  
       REGIAO.ID_REGIAO 
 HAVING 
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) > 0;  
```


Emissão de CO2 produzidas por combústiveis fósseis anual por país (Tabela: emissao_co2_per_capita)

```python
%sql
INSERT INTO workspace.pucrio.emissao_co2_per_capita(id_ano, id_pais, id_fonte_dados, valor_emissao)
SELECT ANO.ID_ANO,  
       PAIS.ID_PAIS,
       23,
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) / populacao_pais.POPULACAO_PAIS VALOR
  FROM  workspace.pucrio.owd_cumulative_co_emissions 
 INNER JOIN workspace.pucrio.ano 
    ON ANO.ANO = owd_cumulative_co_emissions.YEAR  
 INNER JOIN workspace.pucrio.pais
    ON UPPER(PAIS.sigla_pais) = UPPER(owd_cumulative_co_emissions.Code)
 INNER JOIN workspace.pucrio.populacao_pais
    ON populacao_pais.ID_POPULACAO_PAIS = pais.id_pais
   AND populacao_pais.ID_ANO = ano.id_ano 
 WHERE CODE IS NOT NULL 
   AND owd_cumulative_co_emissions.Year > 1950   
 GROUP BY
       ANO.ID_ANO,  
       PAIS.ID_PAIS,
       populacao_pais.POPULACAO_PAIS
 HAVING 
       SUM(owd_cumulative_co_emissions.`Cumulative CO₂ emissions`) / populacao_pais.POPULACAO_PAIS > 0;  
```

##9.1. Emissão CO2 Anual x Fonte de dados.
**Tabela: emissao_co2_anual**

###9.1.1. Detalhes
1. A fonte de dados de: Carbon Monitor é com base nas emissões de queima de combustíveis fósseis e produção de cimento.
2. A fonte de dados de: FAOSTAT é com base nas emissões de carbono por demastamento e queimadas.
3. A fonte de dados de: Our World Data é com base nas emissões de carbono produzidas por combustíveis fósseis e indústria.
4. A fonte de dados da: ONU é com base em emissão CO2 com LULUCF, ou seja, se um país (ano/região/setor/produto) tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, então o valor líquido pode ser negativo.
5. Os dados forma emitidos maior que o ano 2000 (século atual)
6. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.


###9.1.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  a.ano,
  f.nome_fonte,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual ea ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f ON ea.id_fonte_dados = f.id_fonte
WHERE a.ano >= 2000 and a.ano <= 2024
ORDER BY a.ano
```

### 9.1.3. Código Python para geração de gráficos.

```python
import plotly.express as px
import plotly.graph_objects as go
from pyspark.sql import functions as F

def run_query(sql):
    return spark.sql(sql)

df = run_query("""
SELECT DISTINCT
  a.ano,
  f.nome_fonte,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual ea ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f ON ea.id_fonte_dados = f.id_fonte
WHERE a.ano >= 2000 AND A.ANO <= 2024 
ORDER BY a.ano""").toPandas()

# ✅ Gráfico de barras: uma barra por fonte de dados
fig = px.bar(
  df,
  x='ano',
  y='emissao_ton',
  color='nome_fonte',
  barmode='group',  # ← ESSE É O SEGREDO: colunas agrupadas por ano
  title='Emissões Globais de CO₂ por Fonte de Dados (> 2000)',
  labels={
      'ano': 'Ano',
      'emissao_ton': 'Emissões (toneladas)',
      'nome_fonte': 'Fonte de dados'
  },
  color_discrete_sequence=px.colors.qualitative.Bold,
  height=600
)
fig.update_layout(
  xaxis=dict(tickmode='linear', dtick=5),  # mostrar ano a cada 5 anos
  yaxis=dict(title='Emissões (toneladas)'),
  legend_title='Fonte de dados',
  hovermode='x unified'  # ao passar mouse, mostra todas as fontes do mesmo ano
)

fig.show()
```

###9.1.4. Conclusão
Apesar de haver apenas 4 fontes de dados e ainda haver várias formas de cálcular todos os dados de emissões de dióxido de Carbono (CO2), ainda é possível visualizar o seguinte fatos:

Com os dados da ONU desde o ano 2000 até o ano de 2020, é possível visualizar que houve uma diminuição absorção de carbono (o que é ruiim), ou seja, se visualizar os dados de FAOSTAT e cruzar com os dados da ONU, é possível identificar que as emissões de carbono por desmatamento e queimadas (FAOSTAT) continuam crescendo ano após ano e diminuindo a absorção de carbono na atmosfera e aumentando o efeito estufa e piorando o meio ambiente.

Já para os dados de Our World Data (OWD), é possível identificar que o planeta está aumentando a quantidade de emissão de dióxido de carbono por uso de combustíveis fósseis e industrialização, ou seja, mesmo com todas as tecnologias e todos os recursos que são investidos anualmente (na casa dos trilhões de dólares), ainda não conseguimos contornar as situações e diminuir essa quantidade de emissão, piorando ano após ano a qualidade de vida no planeta. 

##9.2. Emissão CO2 Anual x País x Fonte de dados
**Tabela: emissao_co2_anual_pais**

###9.2.1. Detalhes
1. A fonte de dados de: Carbon Monitor é com base nas emissões de queima de combustíveis fósseis e produção de cimento.
2. A fonte de dados de: FAOSTAT é com base nas emissões de carbono por demastamento e queimadas.
3. A fonte de dados de: Our World Data é com base nas emissões de carbono produzidas por combbustíveis fósseis e indústria.
4. A fonte de dados da: ONU é com base em emissão CO2 com LULUCF, ou seja, se um país (ano/região/setor/produto) tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, então o valor líquido pode ser negativo.
5. A fonta de dados da Climate Watch é com base na emissão de CO2 geradas na agricultura com base na emissão total de diversos produtos e também no consumo.
5. Os dados forma emitidos maior que o ano 2000 (século atual)
6. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.

###9.2.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  a.ano,
  f.nome_fonte,
  p.nome_pais,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual_pais ea ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f ON ea.id_fonte_dados = f.id_fonte
INNER JOIN workspace.pucrio.pais p on ea.id_pais = p.id_pais
WHERE a.ano >= 2000 and a.ano <= 2024
ORDER BY a.ano,   p.nome_pais
```

###9.2.3. Código Python para geração de gráficos

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

# 🔹 Executa sua consulta (com todos os países e fontes)
df = spark.sql("""
SELECT DISTINCT
  a.ano,
  f.nome_fonte,
  p.nome_pais,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual_pais ea 
  ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f 
  ON ea.id_fonte_dados = f.id_fonte
INNER JOIN workspace.pucrio.pais p 
  ON ea.id_pais = p.id_pais
WHERE a.ano >= 2000 and a.ano <= 2024
ORDER BY f.nome_fonte, p.nome_pais, a.ano
""").toPandas()

# 🔹 Verifica se há dados
if df.empty:
    raise ValueError("Nenhum dado retornado. Verifique as tabelas e filtros.")

# 🔹 Lista de fontes únicas
fontes = df['nome_fonte'].unique()
print(f"✅ Fontes encontradas ({len(fontes)}): {list(fontes)}")
print(f"✅ Países únicos: {df['nome_pais'].nunique()}")

# 🔹 Gera um gráfico interativo *por fonte*
for fonte in fontes:
    df_fonte = df[df['nome_fonte'] == fonte].copy()
    
    # Limitar a ~20 países para desempenho (ajuste conforme necessidade)
    top_paises = (
        df_fonte.groupby('nome_pais')['emissao_ton']
        .sum()
        .nlargest(20)
        .index
    )
    df_plot = df_fonte[df_fonte['nome_pais'].isin(top_paises)]
    
    # ✅ Plotly Express — interativo por default
    fig = px.line(
        df_plot,
        x='ano',
        y='emissao_ton',
        color='nome_pais',
        title=f'<b>Emissões de CO₂ por País — Fonte: {fonte}</b>',
        labels={
            'ano': 'Ano',
            'emissao_ton': 'Emissões (toneladas)',
            'nome_pais': 'País'
        },
        line_shape='spline',  # suaviza curvas
        markers=True,
        height=700
    )
    
    # Melhorias interativas
    fig.update_layout(
        hovermode='x unified',  # mostra todos os países no mesmo ano ao passar o mouse
        legend=dict(
            title="País",
            orientation="h",
            yanchor="bottom",
            y=-0.3,
            xanchor="center",
            x=0.5,
            font=dict(size=10)
        ),
        xaxis=dict(dtick=2),  # ticks a cada 2 anos
        yaxis=dict(title='Emissões (toneladas)'),
        title_x=0.5
    )
    
    # Exibe o gráfico
    print(f"\n📊 Fonte: {fonte}")
    fig.show()
```

###9.2.4. Conclusão
Na análise um pouco mais detalhada por países, mais especificamente no gráfico da fonte de dados: OWD_ANNUAL_CO_2_EMISSIONS_PER_COUNTRY, é possível visualizar que a China é um país que desde o ano de 2000 está em constante aumento de emissão de dióxido de carbono (CO2). Isso se justifica porque a China era um país que até 1978 estava com o seu comércio e indústria totalmente fechada, também era um país que não constava nas listas das maiores potências mundiais, isso porque após esse ano a China cresceu exponencialmente, onde hoje é a segunda maior potência do planeta. Com isso a China teve que submeter a produzir mais para ter mais investimentos e ter mais recursos, sacrificando mais recursos naturais e emitindo mais dióxido de carbono.

Estados Unidos e Brasil já aparecem com emissões sendo reduzidas em ano após ano ou ainda aumentando a quantidade de emissões anuais bem mais reduzidas, veja:
* Brasil com um aumento de apenas 42% em 24 anos
* Estados Unidos com uma diminuição de 18% nesses mesmos 24 anos
* China com um aumento de incriveis 237% nesses mesmos 24 anos.

Já os dados para a agricultura o valor é ao contrário, há uma baixa no decorrer dos anos, isso ocorre porque os investimentos em novas tecnologias são sempre muito grandes nesse setor, e isso acaba gerano mais produção e menos impacto ambiental.

##9.3. Emissão CO2 Anual x Fonte de dados x PIB.
**Tabela: emissao_co2_anual_pib**

###9.3.1. Detalhes
1. A fonte de dados de: Carbon Monitor é com base nas emissões de queima de combustíveis fósseis e produção de cimento.
2. A fonte de dados de: FAOSTAT é com base nas emissões de carbono por demastamento e queimadas.
3. A fonte de dados de: Our World Data é com base nas emissões de carbono produzidas por combbustíveis fósseis e indústria.
4. A fonte de dados da: ONU é com base em emissão CO2 com LULUCF, ou seja, se um país (ano/região/setor/produto) tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, então o valor líquido pode ser negativo.
5. Os dados forma emitidos maior que o ano 2000 (século atual)
6. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.

###9.3.2. Consulta SQL

```python
%sql
SELECT 
  a.ano, 
  p.nome_pais,
  f.nome_fonte,
  SUM(COALESCE(ea.valor_emissao, 0)) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_pib ea
INNER JOIN workspace.pucrio.ano a ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f ON ea.id_fonte_dados = f.id_fonte
INNER JOIN workspace.pucrio.pais p ON ea.id_pais = p.id_pais
WHERE a.ano BETWEEN 2000 AND 2024       
GROUP BY a.ano, 
  p.nome_pais,
  f.nome_fonte
ORDER BY f.nome_fonte, P.NOME_PAIS, a.ano;
```

### 9.3.3. Código Python para geração de gráficos.

```python
import plotly.express as px
import plotly.graph_objects as go

# Carregar dados
df = spark.sql("""SELECT 
  a.ano, 
  p.nome_pais,
  f.nome_fonte,
  SUM(COALESCE(ea.valor_emissao, 0)) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_pib ea
INNER JOIN workspace.pucrio.ano a ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.catalogo_fontes_dados f ON ea.id_fonte_dados = f.id_fonte
INNER JOIN workspace.pucrio.pais p ON ea.id_pais = p.id_pais
WHERE a.ano BETWEEN 2000 AND 2024       
GROUP BY a.ano, 
  p.nome_pais,
  f.nome_fonte
ORDER BY f.nome_fonte, P.NOME_PAIS, a.ano
""").toPandas()

# 🔹 Verificar se há dados
if df.empty:
    raise ValueError("⚠️ Nenhum dado retornado. Verifique filtros e existência de registros.")

print(f"✅ Dados carregados: {df.shape[0]} linhas | Fontes: {df['nome_fonte'].nunique()} | Países: {df['nome_pais'].nunique()}")

# Limitar a países com pelo menos 10 anos de dados (opcional, para performance)
paises_validos = df.groupby('nome_pais').filter(lambda x: len(x) >= 10)['nome_pais'].unique()

df_filtrado = df[df['nome_pais'].isin(paises_validos)]

# Gerar 1 gráfico por fonte (interativo, com todos os países)
for fonte in df_filtrado['nome_fonte'].unique():
    df_fonte = df_filtrado[df_filtrado['nome_fonte'] == fonte]
    
    fig = px.line(
        df_fonte,
        x='ano',
        y='emissao_ton',
        color='nome_pais',
        title=f'<b>Emissões de CO₂ Anual x Fonte de dados x PIB — Fonte: {fonte}</b>',
        labels={
            'ano': 'Ano',
            'emissao_ton': 'Emissões (toneladas)',
            'nome_pais': 'País'
        },
        height=700,
        markers=True
    )
    
    fig.update_layout(
        hovermode='x unified',
        legend_title="País",
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=-0.3,
            xanchor="center",
            x=0.5
        )
    )
    
    fig.show()
```

###9.3.4. Conclusão
Nestes dados é possível visualizar os dados de emissão de CO2 para os mesmos setores do país pelo PIB deste país, isso faz com que se o PIB do país subiu, como é o caso da China que cresceu exponencialmente ao longo dos anos, o que ocorre é que para o país crescer economicamente deverá investir em mais produção, seja agrícola ou industrial, e para isso gera mais emissões de dióxido de carbono (CO2) e menos absorção por florestas. 

##9.4. Emissão CO2 Anual x Produto
**Tabela: emissao_co2_anual_produto**

###9.4.1. Detalhes
1. A fonte de dados de Climate Watch é para a intensidade de emissões de CO2 agrícolas por produtos produzidos na agricultura e formas diferentes de manejo.
2. Os dados forma emitidos maior que o ano 2000 (século atual)
3. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.

5. Descartado os valores de outras fontes de dados, pois não estão completas o suficiente ou que estão contidos outras emissões de outros gases.

###9.4.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  a.ano,
  P.descricao_produto, 
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual_produto ea ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.produto p ON EA.id_produto = P.id_produto
WHERE a.ano >= 2000 and a.ano <= 2024
AND ea.id_fonte_dados = 3
ORDER BY a.ano, P.descricao_produto
```

###9.4.3 Código Python para a geração de gráficos

```python
import plotly.express as px
import plotly.graph_objects as go
from pyspark.sql import SparkSession

# Carregar dados com a consulta fornecida
df = spark.sql("""
SELECT DISTINCT
  a.ano,
  p.descricao_produto,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.ano a
INNER JOIN workspace.pucrio.emissao_co2_anual_produto ea 
  ON a.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.produto p 
  ON ea.id_produto = p.id_produto
WHERE a.ano BETWEEN 2000 AND 2024
  AND ea.id_fonte_dados = 3  -- Climate Watch Agriculture
ORDER BY a.ano, p.descricao_produto
""").toPandas()

# Verificação de dados
if df.empty:
    raise ValueError("⚠️ Nenhum dado encontrado. Verifique id_fonte_dados=3 e intervalo de anos.")

print(f"✅ Dados carregados: {df.shape[0]} registros | Produtos únicos: {df['descricao_produto'].nunique()}")

# Filtrar apenas produtos com pelo menos 5 anos de dados (evita ruído)
produtos_validos = (
    df.groupby('descricao_produto')
    .filter(lambda x: len(x) >= 5)['descricao_produto']
    .unique()
)
df_plot = df[df['descricao_produto'].isin(produtos_validos)]

#  Gráfico interativo com Plotly
fig = px.line(
    df_plot,
    x='ano',
    y='emissao_ton',
    color='descricao_produto',
    title='<b>Emissões Agrícolas Anuais por Produto (Climate Watch, ID=3)</b>',
    labels={
        'ano': 'Ano',
        'emissao_ton': 'Emissões (toneladas)',
        'descricao_produto': 'Produto Agrícola'
    },
    markers=True,
    height=600  # ← controle de altura para não extrapolar
)

# Melhorias de layout (evita extrapolação + usabilidade)
fig.update_layout(
    hovermode='x unified',  # mostra todos os produtos no mesmo ano
    legend=dict(
        title="Produto",
        orientation="h",
        yanchor="bottom",
        y=-0.25,   # posição logo abaixo do gráfico
        xanchor="center",
        x=0.5,
        font=dict(size=10)
    ),
    xaxis=dict(
        dtick=2,  # ticks a cada 2 anos
        gridcolor='lightgray',
        showgrid=True
    ),
    yaxis=dict(
        title='Emissões (toneladas)',
        gridcolor='lightgray',
        showgrid=True
    ),
    plot_bgcolor='white',
    margin=dict(l=60, r=40, t=60, b=100)
)

# Mostrar
fig.show()
```

###9.4.4. Conclusão
A conversão líquida de florestas tem uma diminuição significativa ao decorrer dos anos, isto é, menos absorção de dióxido de carbono e mais emissão de dióxido de carbono. Pode-se ver as emissões diretas por fertilizantes que tem aumentos significativos, os fertilizantes e agrotóxicos são extremamente nocivos ao meio ambiente, isso porque destroem diretamente a flora e a fauna das suas proximidades de uso, além ainda dos incêndios, que mesmo controlados são extramente prejudiciais ao meio ambiente.

Quanto a emissões provenientes de energia há várias variações ao decorrer dos anos, sendo carvão, eletricidade, gás natural e petroléo, onde há anos que um emite mais que outro em outros anos ao contrário, as quantidades são bem parecidas também.

Já para as informações de produtos provenientes de consumo de carne, tem-se o aumento de emissões por carne bovina e búfalo nos últimos anos e maior consumo sendo o de cabra e um aumento considerável para carne de frango. 

Esses aumentos de emissões por fontes agrícolas fazem sentido quando cruzamos esses dados com a base populacional mundial, pois há um aumento na população anualmente, e isso exige automaticamente, mais comida igual mais emissão de dióxido de carbono (CO2).

3. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.

###9.5.2. Consulta SQL

```python
%sql
SELECT 
  a.ano,
  r.nome_regiao,
  SUM(COALESCE(ea.valor_emissao, 0)) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_regiao ea
JOIN workspace.pucrio.ano a ON ea.id_ano = a.id_ano
JOIN workspace.pucrio.regiao r ON ea.id_regiao = r.id_regiao
WHERE a.ano BETWEEN 2000 AND 2023
GROUP BY a.ano, r.nome_regiao
ORDER BY a.ano, r.nome_regiao
```

###9.5.3. Código Python para geração de gráficos.

```python
import plotly.express as px

# Use a Opção 1 (soma) ou Opção 2 (fonte única) acima
df = spark.sql("""
SELECT 
  a.ano,
  r.nome_regiao,
  SUM(COALESCE(ea.valor_emissao, 0)) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_regiao ea
JOIN workspace.pucrio.ano a ON ea.id_ano = a.id_ano
JOIN workspace.pucrio.regiao r ON ea.id_regiao = r.id_regiao
WHERE a.ano BETWEEN 2000 AND 2023
GROUP BY a.ano, r.nome_regiao
ORDER BY a.ano, r.nome_regiao
""").toPandas()

# Gráfico interativo: linhas por região, eixo X = ano, eixo Y = emissão
fig = px.line(
    df,
    x='ano',
    y='emissao_ton',
    color='nome_regiao',
    title='<b>Emissões de CO₂ por Região (2000–2024)</b>',
    labels={
        'ano': 'Ano',
        'emissao_ton': 'Emissões (toneladas)',
        'nome_regiao': 'Região'
    },
    markers=True,
    height=600
)

fig.update_layout(
    hovermode='x unified',
    legend_title='Região',
    legend=dict(orientation="h", yanchor="bottom", y=-0.25, xanchor="center", x=0.5)
)

fig.show()
```

###9.5.4. Conclusão
A Ásia tendo o maior aumento de emissões nos últimos anos devido à seu aumento populacional e concentração em aumento de PIB (economia), já mostrados em outras fontes de dados, entretanto, todas as regiões aumentaram o consumo de dióxido de carbono (CO2). 



##9.6. Emissão CO2 Anual x Setor

###9.6.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  ea.id_ano,  
  s.descricao_setor,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_setor ea
INNER JOIN workspace.pucrio.setor s ON ea.id_setor = s.id_setor 
WHERE ea.id_ano >= 2000 and ea.id_ano <= 2023
ORDER BY ea.id_ano, s.descricao_setor
```

```python
import plotly.express as px

df = spark.sql("""
SELECT DISTINCT
  ea.id_ano,  
  s.descricao_setor,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_anual_setor ea
INNER JOIN workspace.pucrio.setor s ON ea.id_setor = s.id_setor 
WHERE ea.id_ano >= 2000 and ea.id_ano <= 2021
ORDER BY ea.id_ano, s.descricao_setor
""").toPandas()

fig = px.line(
    df,
    x='id_ano',
    y='emissao_ton',
    color='descricao_setor',
    title='<b>Emissão CO2 Anual x Setor (2000–2021)</b>',
    labels={
        'id_ano': 'Ano',
        'emissao_ton': 'Emissões (toneladas)',
        'descricao_setor': 'Setor'
    },
    markers=True,
    height=600
)

fig.update_layout(
    hovermode='x unified',
    legend_title='Setor',
    legend=dict(orientation="h", yanchor="bottom", y=-0.25, xanchor="center", x=0.5)
)

fig.show()
```

É notável que a maior fonte de emissão está em emissões do setor energetico, através de eletricidade, calor ou combustão estacionária, e em seguida a emissão através de processos industriais (não energéticos), e nesses 2 setores, observa-se que há um aumento de bem significativo ao decorrer dos anos. 

A explicação para esse aumento de emissões de dióxido de carbono ao longo dos anos nesses 2 setores em específico, um é uso de energia e outro é uso industrial, ou seja, quanto mais consumo no planeta maior será a produção industrial e maior será o uso de energia, e com isso consequentemente o aumento da emissão dióxido de carbono. O consumo de aparelhos elétricos e eletrônicos, não provinientes da indústria, mas sim pelas pessoas que também é contabilizado para a emissão de CO2.



##9.7. Emissão CO2 País x Produto
**Tabela: emissao_co2_pais_produto**

###9.7.1. Detalhes
1. A fonte de dados de: FAOSTAT é com base nas emissões de carbono por demastamento e queimadas. 
2. 1 tonelada de CO2e (dióxido de carbono equivalente) é o padrão para um crédito de carbono.
3. Os valores maiores que 1000 e abaixo de -1000 diferentes disso serão ignorados, isso para não sobrecarregar tanto o gráfico e também para não ignorar os valores que estão abaixo de zero que são de terrenos florestais que não emitem CO2 e absorvem CO2
4. Foram considerados apenas os países das regiões: AMERICA DO NORTE, AMERICA DO SUL, AMERICA CENTRAL, EUROPA e ASIA

###9.7.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  A.NOME_PAIS,
  P.descricao_produto,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_pais_produto ea
INNER JOIN workspace.pucrio.produto p ON ea.id_produto = p.id_produto
INNER JOIN workspace.pucrio.pais a ON ea.id_pais = a.id_pais 
INNER JOIN workspace.pucrio.regiao R ON a.id_regiao = R.id_regiao
WHERE (EA.valor_emissao > 1000 OR EA.valor_emissao < -1000)
  AND R.id_regiao IN (4,5,6,7,9)
ORDER BY A.NOME_PAIS,
  P.descricao_produto
```

### 9.7.3. Código Python para geração de gráficos.

```python
import plotly.express as px
import plotly.graph_objects as go
 
df = spark.sql("""
SELECT DISTINCT
  A.NOME_PAIS,
  P.descricao_produto,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_pais_produto ea
INNER JOIN workspace.pucrio.produto p ON ea.id_produto = p.id_produto
INNER JOIN workspace.pucrio.pais a ON ea.id_pais = a.id_pais 
INNER JOIN workspace.pucrio.regiao R ON a.id_regiao = R.id_regiao
WHERE (EA.valor_emissao > 1000 OR EA.valor_emissao < -1000)
  AND R.id_regiao IN (4,5,6,7,9)
ORDER BY A.NOME_PAIS,
  P.descricao_produto
""").toPandas()
 
if df.empty:
    raise ValueError("⚠️ Nenhum dado retornado. Verifique as tabelas emissao_co2_anual_setor, ano e setor.")

print(f"✅ Dados: {df.shape[0]} registros | Produtos: {df['descricao_produto'].nunique()} | Países: {df['NOME_PAIS'].nunique()}")
 
fig = px.bar(
    df,
    x='NOME_PAIS',
    y='emissao_ton',
    color='descricao_produto',
    barmode='group',  
    title='<b>Emissão CO2 País x Produto</b>',
    labels={
        'NOME_PAIS': 'País',
        'emissao_ton': 'Emissões (toneladas)',
        'descricao_produto': 'Produto'
    },
    height=650,
    text_auto=True   
)
 
fig.update_layout(
    hovermode='x unified',  
    legend_title='Produto',
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=-0.25,
        xanchor='center',
        x=0.5,
        font=dict(size=10)
    ),
    xaxis=dict(
        dtick=2,   
        tickmode='linear'
    ),
    yaxis=dict(
        title='Emissões (toneladas)',
        tickformat=',.0f'  
    ),
    margin=dict(l=60, r=40, t=70, b=120)
)
 
fig.show()
```

###9.7.4. Conclusão
As informações desta fonte de dados podem ser cruzadas com outras fontes de dados foram apenas catalogadas, afinal ela mostra um dado fato interessantissímo, que a Russia é o maior em absorção de dióxido de carbono (CO2), entretanto, sabe-se que a maior floresta e maior reserva natural do mundo fica no Brasil, que é a Amazônia, sendo que o Brasil neste ranking aparece em 4º lugar, atrás de China seguido pelo Estados Unidos. 

Mas isso gera novas perguntas, como a Rússia conseguiu absorver mais dióxido de carbono do que o Brasil, sendo que há mais florestas, e a resposta está nos dados da NASA. Se observar as latitudes e longitudes obtidos da NASA, verá que tem alguns que são relacionados à Antartida, e lá não há vida humana e nem florestas que absorvem dióxido de carbono, apenas geleiras e alguns poucos animais, a emissão é relativamente mínima. 

E com essa informação é possível chegar à conclusão de que a Rússia possuí muitas florestas e mais importante é que possuí muito gelo durante a maior parte do ano, e conforme dados científicos, o gelo absorve e reflete os raios ultravioletas e transforma a emissão do dióxido de carbono em absorção disso, diminuindo assim o efeito estufa.

##9.9. Emissão CO2 Per Capita.
**Tabela: emissao_co2_per_capita**


###9.9.1 Detalhes
1. A fonte de dados de: Carbon Monitor é com base nas emissões de queima de combustíveis fósseis e produção de cimento.
2. A fonte de dados de: FAOSTAT é com base nas emissões de carbono por demastamento e queimadas.
3. A fonte de dados de: Our World Data é com base nas emissões de carbono produzidas por combbustíveis fósseis e indústria.
4. A fonte de dados da: ONU é com base em emissão CO2 com LULUCF, ou seja, se um país (ano/região/setor/produto) tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, então o valor líquido pode ser negativo.
5. A fonte de dados de: Climate Watch é o histórico referente à quantidade de emissões de dióxido de carbono.
6. Os dados forma emitidos maior que o ano 2000 (século atual)

###9.9.2. Consulta SQL

```python
%sql
SELECT DISTINCT
  s.ano,
  p.nome_pais,
  F.NOME_FONTE,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_per_capita ea
INNER JOIN workspace.pucrio.ano s ON s.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.pais p ON ea.id_pais = p.id_pais
INNER JOIN workspace.PUCRIO.catalogo_fontes_dados F ON ea.id_fonte_dados = F.id_fonte
WHERE EA.valor_emissao > 0
AND S.ANO >= 2000 AND S.ANO <= 2024
ORDER BY F.NOME_FONTE, p.nome_pais, s.ano
```

### 9.9.3. Código Python para geração de gráficos.

```python
import plotly.express as px
import pandas as pd

df = spark.sql("""
SELECT DISTINCT
  s.ano,
  p.nome_pais,
  F.NOME_FONTE,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_per_capita ea
INNER JOIN workspace.pucrio.ano s ON s.id_ano = ea.id_ano
INNER JOIN workspace.pucrio.pais p ON ea.id_pais = p.id_pais
INNER JOIN workspace.pucrio.catalogo_fontes_dados F ON ea.id_fonte_dados = F.id_fonte
WHERE ea.valor_emissao > 0
  AND s.ano BETWEEN 2000 AND 2024
ORDER BY F.NOME_FONTE, p.nome_pais, s.ano
""").toPandas()

print(f"✅ Dados carregados: {df.shape[0]} registros")
print(f"✅ Países únicos: {df['nome_pais'].nunique()}")
print(f"✅ Fontes únicas: {df['NOME_FONTE'].nunique()}")
print(f"✅ Anos: {sorted(df['ano'].unique())}")

for fonte in df['NOME_FONTE'].unique():
    df_fonte = df[df['NOME_FONTE'] == fonte]
    
    fig = px.line(
        df_fonte,
        x='ano',
        y='emissao_ton',
        color='nome_pais',
        title=f'<b>Emissões per capita por País — Fonte: {fonte}</b>',
        labels={
            'ano': 'Ano',
            'emissao_ton': 'Emissões per capita (toneladas)',
            'nome_pais': 'País'
        },
        height=700,
        markers=True
    )
    
    fig.update_layout(
        hovermode='x unified', 
        legend=dict(
            title="País",
            orientation="v",          
            yanchor="top",
            y=1.0,
            xanchor="left",
            x=1.02,
            font=dict(size=9),
            itemsizing="constant"
        ),
        xaxis=dict(dtick=2),
        yaxis=dict(title='Toneladas CO₂ / pessoa'),
        margin=dict(l=60, r=220, t=70, b=60)  
    )
    
    print(f"\n📊 Fonte: {fonte}")
    fig.show()
```

###9.9.4. Conclusão
Para esta conclusão será considerada apenas os gráficos e dados de FAOSTAT e OWD, que estão com informações mais completas sobre os países e as emissõeo per capita. 

Para a FAOSTAT, que são as informações de emissão de dióxido de carbono por desmatamento e queimadas, para fins agrícolas mostra um ponto importante que o Brasil é o país com mais queimadas e desmatamentos para criar novas áreas agrícolas, é praticamente mais que o dobro dos outros países, se juntar China e Russia quase não alcança a quantidade do Brasil, isso é um dado muito ruim, afinal, é um país que produz muito na agricultura, mas desmata e queima muito gerando muitos impactos ambientais anuais, o lado bom é que desde esse valor vem caindo ano após ano, tendo um aumento considerável em 2022 e mantendo-se no mesmo nível até 2024.

Já as informações de OWD, que são com base de combustíveis fósseis e indústrial, Estados Unidos sempre liderou esse ranking, porém vem caindo ao decorrer dos anos, já a Rússia e China vem crescendo exponencialmente nesse valor e o Brasil mantendo-se em uma linha bem abaixo desses países, afinal o Brasil não teve grandes aumentos indústriais, mas teve aumento de consumo de combustíveis fósseis, mas por outro lado a população cresceu na mesma proporção.

###9.10.1. Detalhes

###9.13.2. Consulta SQL

```python
%sql
SELECT DISTINCT  
  s.descricao_setor, 
  F.NOME_FONTE,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_setor ea
INNER JOIN workspace.PUCRIO.catalogo_fontes_dados F ON ea.id_fonte_dados = F.id_fonte
INNER JOIN workspace.pucrio.setor s ON ea.id_setor = s.id_setor
WHERE EA.valor_emissao <> 0 
ORDER BY F.NOME_FONTE,  
  s.descricao_setor
```

###9.13.3. Script Python para a geração de gráficos

```python
import plotly.express as px
import plotly.graph_objects as go
import textwrap
import pandas as pd
 
df = spark.sql("""
SELECT DISTINCT  
  s.descricao_setor, 
  F.NOME_FONTE,
  COALESCE(ea.valor_emissao, 0) AS emissao_ton
FROM workspace.pucrio.emissao_co2_setor ea
INNER JOIN workspace.pucrio.catalogo_fontes_dados F 
  ON ea.id_fonte_dados = F.id_fonte
INNER JOIN workspace.pucrio.setor s 
  ON ea.id_setor = s.id_setor
WHERE ea.valor_emissao <> 0 
ORDER BY F.NOME_FONTE, s.descricao_setor
""").toPandas()
 
if df.empty:
    raise ValueError("⚠️ Nenhum dado retornado. Verifique emissao_co2_setor e filtros.")
print(f"✅ Dados carregados: {df.shape[0]} registros | Fontes: {df['NOME_FONTE'].nunique()} | Setores: {df['descricao_setor'].nunique()}")
 
def wrap_text(text, width=18):
    return "<br>".join(textwrap.wrap(str(text), width=width))

df['NOME_FONTE_wrapped'] = df['NOME_FONTE'].apply(wrap_text)
df['descricao_setor_wrapped'] = df['descricao_setor'].apply(lambda x: wrap_text(x, width=15))
 
fig = px.bar(
    df,
    x='NOME_FONTE_wrapped',
    y='emissao_ton',
    color='descricao_setor_wrapped',
    barmode='group',
    title='<b>Emissões Totais por Setor — Comparação entre Fontes de Dados</b>',
    labels={
        'NOME_FONTE_wrapped': 'Fonte de Dados',
        'emissao_ton': 'Emissões (toneladas)',
        'descricao_setor_wrapped': 'Setor'
    },
    height=650,
    text_auto=True  
)
 
fig.update_layout(
    hovermode='x unified',   
    legend=dict(
        title="Setor",
        orientation='v',
        yanchor='top',
        y=1.0,
        xanchor='left',
        x=1.04,
        font=dict(size=9),
        itemwidth=120,
        itemsizing="constant"
    ),
    xaxis=dict(
        tickangle=0,  
        title='Fonte de Dados'
    ),
    yaxis=dict(
        title='Emissões (toneladas)',
        tickformat=',.0f',
        rangemode='tozero'   
    ),
    margin=dict(l=60, r=200, t=70, b=100)
)
 
fig.show()
```

###9.13.4. Conclusão
Emissões do setor de energia tem o maior índice disparado de todos os demais setores, entretanto um outro setor destaca-se muito que é o setor de processos indústriais não energéticos, a explicação é a seguinte: 

_"Processos industriais não energéticos são atividades de transformação química e/ou física de matérias-primas que geram emissões atmosféricas, mas que não estão diretamente relacionadas à queima de combustíveis para a produção de energia ou calor. As emissões resultantes são um subproduto intrínseco da reação química ou do processo de fabricação em si."_

Em resumo são emissões altissímas de produção de cimento, ferro, aço, química de ácidos (súlfurico, nítrico, etc), alumínio. Estes materiais são usados em quase todos os demais setores, e para as mais diversas finalidades.

##10.1. Qual a emissão de carbono no planeta por ano?
**Tabela: emissao_co2_anual** 

**Fontes de dados (descrição - nome - id):** 
* Organização das Nações Unidas (ONU) - Emissões totais de CO2 por ano - ONU_TIME_SERIES_CO_2_TOTAL - 17
* Our World in Data - Emissões totais acumuladas de CO2 - OWD_CUMULATIVE_CO_EMISSIONS - 23

1. Our World Data é com base nas emissões de carbono produzidas por combustíveis fósseis e indústria.
2. ONU é com base em emissão CO2 com LULUCF, ou seja, se um país (ano/região/setor/produto) tem um grande reflorestamento que absorve mais carbono do que o desmatamento emite, então o valor líquido pode ser negativo.

```python
%sql
SELECT DISTINCT   
       CATALOGO_FONTES_DADOS.NOME_FONTE,
       ANO.ANO,
       ROUND(COALESCE(emissao_co2_anual.valor_emissao, 0), 2) VALOR_EMISSAO 
  FROM workspace.pucrio.emissao_co2_anual
 INNER JOIN workspace.pucrio.ano
    ON ano.id_ano = emissao_co2_anual.id_ano
 INNER JOIN workspace.pucrio.catalogo_fontes_dados
    ON catalogo_fontes_dados.id_fonte = emissao_co2_anual.id_fonte_dados
 WHERE catalogo_fontes_dados.id_fonte IN (17, 23)     
   AND ANO.ANO >= 2000 AND ANO.ANO <= 2024
 ORDER BY 
       catalogo_fontes_dados.NOME_FONTE,
       ANO.ano

        

  


```

```python
import plotly.express as px
import plotly.graph_objects as go
from pyspark.sql import functions as F

def run_query(sql):
    return spark.sql(sql)

df = run_query("""
        SELECT DISTINCT   
            CATALOGO_FONTES_DADOS.NOME_FONTE,
            ANO.ANO,
            ROUND(COALESCE(emissao_co2_anual.valor_emissao, 0), 2) VALOR_EMISSAO 
        FROM workspace.pucrio.emissao_co2_anual
        INNER JOIN workspace.pucrio.ano
            ON ano.id_ano = emissao_co2_anual.id_ano
        INNER JOIN workspace.pucrio.catalogo_fontes_dados
            ON catalogo_fontes_dados.id_fonte = emissao_co2_anual.id_fonte_dados
        WHERE catalogo_fontes_dados.id_fonte IN (17, 23)     
        AND ANO.ANO >= 2000 AND ANO.ANO <= 2024
        ORDER BY 
            catalogo_fontes_dados.NOME_FONTE,
            ANO.ano
    """).toPandas()
 
fig = px.bar(
  df,
  x='ANO',
  y='VALOR_EMISSAO',
  color='NOME_FONTE',
  barmode='group',  
  title='Emissões Globais de CO₂ por Fonte de Dados X Ano (2000 até 2024)',
  labels={
      'ANO': 'Ano',
      'VALOR_EMISSAO': 'Emissões (toneladas)',
      'NOME_FONTE': 'Fonte de dados'
  },
  color_discrete_sequence=px.colors.qualitative.Bold,
  height=600
)
fig.update_layout(
  xaxis=dict(tickmode='linear', dtick=5), 
  yaxis=dict(title='Emissões (toneladas)'),
  legend_title='Fonte de dados',
  hovermode='x unified' 
)

fig.show()
```

Após executar o comando em Python com a consulta, foi gerado dois gráficos com as fontes de dados da ONU e OWD, abaixo as conclusões de cada um o print screen do gráfico gerado para cada.

##10.2. Qual a emissão de carbono no planeta por país x ano?

**Tabela: emissao_co2_anual_pais**

**Fontes de dados (descrição - nome - id):**
* Our World in Data - Emissões totais de CO2 por país	OWD_ANNUAL_CO_2_EMISSIONS_PER_COUNTRY -	18
* Our World in Data - Emissões totais acumuladas de CO2	OWD_CUMULATIVE_CO_EMISSIONS	- 23
* Food And Agriculture Organization of the United Nations - Informações sobre tipos de alimentos, ano e tipos de gases emitidos - FAOSTAT_EMISSIONS_TOTALS - 13


```python
%sql
SELECT DISTINCT     
       CATALOGO_FONTES_DADOS.NOME_FONTE,
       PAIS.nome_pais,
       ANO.ANO,
       SUM(ROUND(COALESCE(emissao_co2_anual_pais.valor_emissao, 0), 2)) VALOR_EMISSAO
  FROM workspace.pucrio.emissao_co2_anual_pais
 INNER JOIN workspace.pucrio.pais
    ON pais.id_pais = emissao_co2_anual_pais.id_pais
 INNER JOIN workspace.pucrio.catalogo_fontes_dados
    ON catalogo_fontes_dados.id_fonte = emissao_co2_anual_pais.id_fonte_dados
 INNER JOIN workspace.pucrio.ano
    ON ano.id_ano = emissao_co2_anual_pais.id_ano
 WHERE 1 = 1  
   AND catalogo_fontes_dados.id_fonte IN (13, 23, 18)   
   AND ANO.ANO >= 2000 AND ANO.ANO <= 2024    
 GROUP BY     
       CATALOGO_FONTES_DADOS.NOME_FONTE,
       ANO.ANO,
       PAIS.nome_pais
 ORDER BY 
       catalogo_fontes_dados.NOME_FONTE,       
       pais.nome_pais,
       ANO.ANO
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

df = spark.sql(""" 
SELECT DISTINCT     
       CATALOGO_FONTES_DADOS.NOME_FONTE,
       PAIS.nome_pais,
       ANO.ANO,
       SUM(ROUND(COALESCE(emissao_co2_anual_pais.valor_emissao, 0), 2)) VALOR_EMISSAO
  FROM workspace.pucrio.emissao_co2_anual_pais
 INNER JOIN workspace.pucrio.pais
    ON pais.id_pais = emissao_co2_anual_pais.id_pais
 INNER JOIN workspace.pucrio.catalogo_fontes_dados
    ON catalogo_fontes_dados.id_fonte = emissao_co2_anual_pais.id_fonte_dados
 INNER JOIN workspace.pucrio.ano
    ON ano.id_ano = emissao_co2_anual_pais.id_ano
 WHERE 1 = 1  
   AND catalogo_fontes_dados.id_fonte IN (13, 23, 18)   
   AND ANO.ANO >= 2000 AND ANO.ANO <= 2024    
 GROUP BY     
       CATALOGO_FONTES_DADOS.NOME_FONTE,
       ANO.ANO,
       PAIS.nome_pais
 ORDER BY 
       catalogo_fontes_dados.NOME_FONTE,       
       pais.nome_pais,
       ANO.ANO
""").toPandas()

if df.empty:
    raise ValueError("Nenhum dado retornado. Verifique as tabelas e filtros.")

fontes = df['NOME_FONTE'].unique()
print(f"✅ Fontes encontradas ({len(fontes)}): {list(fontes)}")
print(f"✅ Países únicos: {df['nome_pais'].nunique()}")

for fonte in fontes:
    df_fonte = df[df['NOME_FONTE'] == fonte].copy()
    
    top_paises = (
        df_fonte.groupby('nome_pais')['VALOR_EMISSAO']
        .sum()
        .nlargest(20)
        .index
    )
    df_plot = df_fonte[df_fonte['nome_pais'].isin(top_paises)]
    
    fig = px.line(
        df_plot,
        x='ANO',
        y='VALOR_EMISSAO',
        color='nome_pais',
        title=f'<b>Emissões de CO₂ por País X Ano — Fonte: {fonte}</b>',
        labels={
            'ANO': 'Ano',
            'emissao_ton': 'Emissões (toneladas)',
            'nome_pais': 'País'
        },
        line_shape='spline', 
        markers=True,
        height=700
    )
    
    fig.update_layout(
        hovermode='x unified',  
        legend=dict(
            title="País",
            orientation="h",
            yanchor="bottom",
            y=-0.3,
            xanchor="center",
            x=0.5,
            font=dict(size=10)
        ),
        xaxis=dict(dtick=2), 
        yaxis=dict(title='Emissões (toneladas)'),
        title_x=0.5
    )
     
    print(f"\n📊 Fonte: {fonte}")
    fig.show()
```

###10.2.1. Conclusão

Foram selecionados alguns países que possuem quantidades signficativas nos gráficos abaixo para todas as fontes de dados, para melhorar a visualização e a explicação de cada um. 

No caso da fonte de FAOSTAT refere-se à emissões de dióxido de carbono desmatamento e queimadas, sendo o país com o maior índice e o mais crescente é a China, devido ao aumento populacional, áreas agrícolas, áreas industriais e entre outros, pois o mesmo ocorre com a Índia. Já o Brasil em alguns anos consegue combater essa estatística e manter abaixo do ano de 2010, entretanto esse número deve aumentar, pois todos os anos as áreas agrícolas aumentam muito e os desmatamentos e queimadas acontecem, mesmo que muitas vezes naturalmente.
