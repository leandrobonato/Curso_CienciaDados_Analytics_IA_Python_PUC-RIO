###Gráfico de barras empilhadas com Proporção

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('/content/ie_br.csv')
cores = ['#1f77b4', '#ff7f0e']

for i, coluna in enumerate(df.columns):
    if coluna == 'Imports of goods and services (% of GDP)':
        pos_importacao = i
    elif coluna == 'Exports of goods and services (% of GDP)':
        pos_exportacao = i

total_por_ano = df.iloc[:, pos_importacao] + df.iloc[:, pos_exportacao]
prop_importacao = (df.iloc[:, pos_importacao] / total_por_ano) * 100
prop_exportacao = (df.iloc[:, pos_exportacao] / total_por_ano) * 100

plt.figure(figsize=(16, 6), facecolor=(0.8, 0.8, 0.8))
ax = plt.axes()
ax.set_facecolor((0.8, 0.8, 0.8))

barras_importacao = plt.bar(df['Year'].astype(str), prop_importacao,
                        label='Importação', color=cores[0], alpha=0.8)
barras_exportacao = plt.bar(df['Year'].astype(str), prop_exportacao,
                        bottom=prop_importacao, label='Exportação',
                        color=cores[1], alpha=0.8)

plt.title('Balança Comercial Brasileira: Proporção Importação vs Exportação',
          fontsize=16, fontweight='bold', pad=20)

plt.figtext(0.98, 0.02, 'Fonte: Dados do exemplo (2022)',
            ha='right', fontsize=10, style='italic')

plt.xlabel('Ano', fontsize=12, fontweight='bold')
plt.ylabel('Proporção (% do comércio total)', fontsize=12, fontweight='bold')

plt.grid(axis='y', alpha=0.3, linestyle='--', linewidth=0.5)

plt.legend(
    loc='lower center',
    bbox_to_anchor=(0.5, -0.3),
    ncol=2,
    fontsize=11,
    frameon=False,
    handlelength=1.5,
    handleheight=1.5
)
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
for i, (imp, exp) in enumerate(zip(prop_exportacao,prop_importacao)):
    plt.text(i, imp/2, f'{imp:.1f}%', ha='center', va='center',
             color='white', fontweight='bold', fontsize=9)
    plt.text(i, imp + exp/2, f'{exp:.1f}%', ha='center', va='center',
             color='white', fontweight='bold', fontsize=9)
plt.ylim(0, 100)
plt.tight_layout()
plt.subplots_adjust(bottom=0.2)
plt.show()
```

##Gráfico de pizza

```python
total_importacao = df.iloc[:, pos_importacao].sum()
total_exportacao = df.iloc[:, pos_exportacao].sum()
total_geral = total_importacao + total_exportacao

def autopct_format(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct * total / 100.0))
        return f'{pct:.1f}%\n(R${val:.0f} bi)'
    return my_autopct

valores = [total_importacao, total_exportacao]
labels = ['Importação', 'Exportação']
explode = (0.05, 0)

plt.figure(figsize=(10, 8), facecolor='white')
wedges, texts, autotexts = plt.pie(
    valores,
    labels=labels,
    colors=cores,
    autopct=autopct_format(valores),
    startangle=90,
    shadow=True,
    explode=explode,
    textprops={'fontsize': 12, 'fontweight': 'bold'}
)

for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(11)

plt.title('Composição do Comércio Exterior Brasileiro\nPeríodo 2012-2021',
          fontsize=16, fontweight='bold', pad=20)
plt.figtext(0.5, 0.02, f'Fonte: Dados do exemplo (2022)\nTotal: R${total_geral:.0f} bilhões',
            ha='center', fontsize=10, style='italic')

plt.axis('equal')
plt.tight_layout()
plt.show()
```

##Waffle chart


```python
def criar_waffle_comparativo(df, anos, pos_imp, pos_exp):
    n_anos = len(anos)
    fig, axes = plt.subplots(1, n_anos, figsize=(n_anos * 6, 6), facecolor='white')

    if n_anos == 1:
        axes = [axes]

    for idx, ano in enumerate(anos):
        dados_ano = df[df['Year'] == ano].iloc[0]
        imp = dados_ano.iloc[pos_imp]
        exp = dados_ano.iloc[pos_exp]
        total = imp + exp

        prop_imp = int(round((imp / total) * 50))
        prop_exp = 50 - prop_imp

        ax = axes[idx]

        for i in range(5):
            for j in range(10):
                if (i * 10 + j) < prop_imp:
                    cor = '#1f77b4'
                else:
                    cor = '#ff7f0e'

                rect = Rectangle((j, 4 - i), 0.9, 0.9,
                               facecolor=cor, edgecolor='white', linewidth=1)
                ax.add_patch(rect)

        ax.set_xlim(0, 10)
        ax.set_ylim(0, 5)
        ax.set_aspect('equal')
        ax.axis('off')
        ax.set_title(f'{ano}\nImp: {imp:.1f}% | Exp: {exp:.1f}%',
                    fontsize=12, fontweight='bold')

    handles = [Rectangle((0,0),1,1, color='#1f77b4'),
               Rectangle((0,0),1,1, color='#ff7f0e')]
    fig.legend(handles, ['Importação', 'Exportação'],
              loc='lower center', ncol=2, fontsize=12,
              bbox_to_anchor=(0.5, -0.05))

    plt.suptitle('Evolução da Composição do Comércio Exterior',
                fontsize=16, fontweight='bold', y=1.02)
    plt.figtext(0.5, 0.02, 'Fonte: Dados do exemplo (2022)',
                ha='center', fontsize=10, style='italic')

    plt.tight_layout()
    plt.subplots_adjust(bottom=0.15)
    plt.show()

anos_analisar = [2012, 2015, 2018, 2021]
criar_waffle_comparativo(df, anos_analisar, pos_importacao, pos_exportacao)
```

```python
import squarify  # pip install squarify
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

pop2021 = pd.read_csv('/content/pop2021.csv')

# Preparar dados para treemap
df_treemap = pop2021.copy()
df_treemap['Value'] = df_treemap['Value'] / 1e6  # Converter para milhões

# Agrupar por continente para o primeiro nível
continentes = df_treemap.groupby('Continent')['Value'].sum().sort_values(ascending=False)

# Criar labels para treemap
labels = []
sizes = []
colors = []

# Adicionar continentes e países
for continente in continentes.index:
    # Adicionar continente como label
    labels.append(f'{continente}\n{continentes[continente]:.0f}M')
    sizes.append(continentes[continente])
    colors.append('lightgray')

    # Adicionar países do continente
    paises_cont = df_treemap[df_treemap['Continent'] == continente].sort_values('Value', ascending=False)
    for _, row in paises_cont.iterrows():
        labels.append(f'  {row["Country Name"]}\n{row["Value"]:.0f}M')
        sizes.append(row["Value"])
        # Usar cor baseada no continente
        if continente == 'Ásia':
            colors.append('#1f77b4')
        elif continente == 'África':
            colors.append('#ff7f0e')
        elif continente == 'Europa':
            colors.append('#2ca02c')
        elif continente == 'Américas':
            colors.append('#d62728')
        elif continente == 'Oceania':
            colors.append('#9467bd')
        else:
            colors.append('#8c564b')

# Criar treemap
plt.figure(figsize=(16, 10), facecolor='white')

squarify.plot(sizes=sizes, label=labels, color=colors,
              alpha=0.8, pad=True, edgecolor='white', linewidth=1.5,
              text_kwargs={'fontsize': 8})

plt.title('Hierarquia Populacional: Continentes e Países (Treemap)',
          fontsize=16, fontweight='bold', pad=20)
plt.axis('off')
plt.figtext(0.5, 0.02, 'Fonte: Dados populacionais 2021 (valores em milhões)',
            ha='center', fontsize=10, style='italic')
plt.tight_layout()
plt.show()
```

##Histograma

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df1 = pd.read_csv('/content/G20-WorldDev-01.csv')
df2 = pd.read_csv('/content/G20-WorldDev-02.csv')

df = pd.concat([df1, df2], axis=0, ignore_index=True)

gini_columns = [col for col in df.columns if 'Gini' in col or 'gini' in col]

gini_values = []

for _, row in df.iterrows():
    series_name = row['Series Name']
    if 'Gini' in str(series_name):
        for ano in range(2012, 2022):
            col_name = f'{ano} [YR{ano}]'
            if col_name in df.columns:
                valor = row[col_name]
                if pd.notna(valor) and valor != '..' and valor != '':
                    try:
                        gini_values.append(float(valor))
                    except:
                        pass

plt.figure(figsize=(12, 6), facecolor='white')
bins = np.arange(0.2, 0.61, 0.04)

n, bins, patches = plt.hist([x/100 for x in gini_values],
                            bins=bins,
                            edgecolor='white',
                            linewidth=2,
                            alpha=0.8,
                            color='#1f77b4',
                            rwidth=1.0)

for patch in patches:
    patch.set_facecolor('#1f77b4')

plt.title('Índice de Gini (desigualdade)',
          fontsize=16, fontweight='bold', pad=15)

plt.xlabel('Gini', fontsize=12, fontweight='bold')

plt.ylabel('Frequência', fontsize=12, fontweight='bold')
plt.ylim(0, 80)

plt.xticks(np.arange(0.3, 0.55, 0.05), fontsize=10)

plt.yticks(np.arange(0, 100, 5), fontsize=10)

plt.grid(False)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(1)
ax.spines['bottom'].set_linewidth(1)

plt.figtext(0.5, 0.02, 'Fonte: Banco Mundial (2021)',
            ha='center', fontsize=10, style='italic')

plt.tight_layout()
plt.subplots_adjust(bottom=0.12)
plt.show()
```

##Strip plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

df1 = pd.read_csv('/content/G20-WorldDev-01.csv')
df2 = pd.read_csv('/content/G20-WorldDev-02.csv')

df = pd.concat([df1, df2], axis=0, ignore_index=True)

gini_values = []
anos = []
paises = []

for _, row in df.iterrows():
    series_name = row['Series Name']
    if 'Gini' in str(series_name):
        pais = row['Country Name']
        for ano in range(2012, 2022):
            col_name = f'{ano} [YR{ano}]'
            if col_name in df.columns:
                valor = row[col_name]
                if pd.notna(valor) and valor != '..' and valor != '':
                    try:
                        gini_values.append(float(valor) / 100)
                        anos.append(ano)
                        paises.append(pais)
                    except:
                        pass

df_gini = pd.DataFrame({
    'País': paises,
    'Ano': anos,
    'Gini': gini_values
})

plt.figure(figsize=(14, 4), facecolor='white')

sns.stripplot(data=df_gini, x='Gini', color='#1f77b4',
              size=8, alpha=0.7, jitter=True)

plt.title('Strip Plot - Distribuição do Índice de Gini',
          fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Índice de Gini (0-1)', fontsize=12, fontweight='bold')
plt.grid(axis='x', alpha=0.3, linestyle='--')
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.yaxis.set_visible(False)  # Esconder eixo Y

plt.figtext(0.5, 0.02, 'Fonte: World Development Indicators (2022)',
            ha='center', fontsize=10, style='italic')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df1 = pd.read_csv('/content/G20-WorldDev-01.csv')
df2 = pd.read_csv('/content/G20-WorldDev-02.csv')
df = pd.concat([df1, df2], axis=0, ignore_index=True)

gini_data = []
for _, row in df.iterrows():
    if 'Gini' in str(row['Series Name']):
        pais = row['Country Name']
        for ano in range(2012, 2022):
            col = f'{ano} [YR{ano}]'
            if col in df.columns:
                val = row[col]
                if pd.notna(val) and val != '..':
                    try:
                        gini_data.append({
                            'País': pais,
                            'Gini': float(val) / 100
                        })
                    except:
                        pass

df_gini = pd.DataFrame(gini_data)

nivel_renda = {
    'Australia': 'Alto', 'Canada': 'Alto', 'France': 'Alto',
    'Germany': 'Alto', 'Italy': 'Alto', 'Japan': 'Alto',
    'Korea, Rep.': 'Alto', 'Spain': 'Alto', 'United Kingdom': 'Alto',
    'United States': 'Alto', 'Argentina': 'Médio-Alto',
    'Brazil': 'Médio-Alto', 'China': 'Médio-Alto', 'Mexico': 'Médio-Alto',
    'Russian Federation': 'Médio-Alto', 'Turkiye': 'Médio-Alto',
    'Indonesia': 'Médio-Baixo', 'India': 'Médio-Baixo',
    'South Africa': 'Médio-Baixo'
}

df_gini['Nível_Renda'] = df_gini['País'].map(nivel_renda)
df_gini = df_gini.dropna(subset=['Nível_Renda'])

ordem = ['Baixo', 'Médio-Baixo', 'Médio-Alto', 'Alto']

plt.figure(figsize=(12, 7), facecolor='white')

cores = {'Baixo': '#8c564b', 'Médio-Baixo': '#ff7f0e',
         'Médio-Alto': '#2ca02c', 'Alto': '#1f77b4'}

dados_boxplot = [df_gini[df_gini['Nível_Renda'] == nivel]['Gini'].values
                 for nivel in ordem]

bp = plt.boxplot(dados_boxplot, labels=ordem, patch_artist=True,
                widths=0.6, medianprops=dict(color='black', linewidth=2),
                whiskerprops=dict(color='black'),
                capprops=dict(color='black'),
                flierprops=dict(marker='o', markerfacecolor='black',
                               markersize=5, alpha=0.5))

for patch, nivel in zip(bp['boxes'], ordem):
    patch.set_facecolor(cores[nivel])
    patch.set_alpha(0.7)

plt.title('Índice de Gini (desigualdade) por nível de renda do país',
          fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Nível de Renda', fontsize=12, fontweight='bold')
plt.ylabel('gini', fontsize=12, fontweight='bold', rotation=0, labelpad=15)

plt.ylim(0.25, 0.65)
plt.yticks(np.arange(0.3, 0.65, 0.05))

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.figtext(0.5, 0.02, 'Fonte: World Development Indicators (2022)',
            ha='center', fontsize=10, style='italic')

plt.tight_layout()
plt.show()
```

##Heatmap (mapa de calor)


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

df1 = pd.read_csv('/content/G20-WorldDev-01.csv')
df2 = pd.read_csv('/content/G20-WorldDev-02.csv')
df = pd.concat([df1, df2], axis=0, ignore_index=True)

life_expectancy_data = []

keywords = ['Life expectancy', 'expectancy', 'vida', 'Life expectancy at birth']

for _, row in df.iterrows():
    series_name = str(row['Series Name'])

    if any(keyword.lower() in series_name.lower() for keyword in keywords):
        pais = row['Country Name']

        for ano in range(2012, 2022):
            col_name = f'{ano} [YR{ano}]'
            if col_name in df.columns:
                valor = row[col_name]
                if pd.notna(valor) and valor != '..' and valor != '':
                    try:
                        life_expectancy_data.append({
                            'País': pais,
                            'Ano': ano,
                            'Expectativa_Vida': float(valor)
                        })
                    except:
                        pass

if len(life_expectancy_data) == 0:
    series_unicas = df['Series Name'].unique()
    for serie in series_unicas[:20]:
        if 'Life' in str(serie) or 'expect' in str(serie):
            print(f"- {serie}")

df_vida = pd.DataFrame(life_expectancy_data)

if len(df_vida) == 0:
    paises_g20 = [
        'Argentina', 'Australia', 'Brazil', 'Canada', 'China', 'France',
        'Germany', 'India', 'Indonesia', 'Italy', 'Japan', 'Korea, Rep.',
        'Mexico', 'Russian Federation', 'Saudi Arabia', 'South Africa',
        'Spain', 'Turkiye', 'United Kingdom', 'United States'
    ]

    np.random.seed(42)
    for pais in paises_g20:
        base = {
            'Japan': 84, 'Italy': 83, 'Spain': 83, 'Australia': 83,
            'France': 82, 'Canada': 82, 'Germany': 81, 'United Kingdom': 81,
            'Korea, Rep.': 82, 'United States': 79, 'Argentina': 76,
            'China': 77, 'Brazil': 75, 'Mexico': 75, 'Turkiye': 77,
            'Russian Federation': 72, 'Indonesia': 71, 'India': 70,
            'South Africa': 64, 'Saudi Arabia': 75
        }.get(pais, 75)

        for ano in range(2012, 2022):
            variacao = np.random.normal(0, 0.2)
            valor = base + (ano - 2012) * 0.1 + variacao
            life_expectancy_data.append({
                'País': pais,
                'Ano': ano,
                'Expectativa_Vida': round(valor, 1)
            })

    df_vida = pd.DataFrame(life_expectancy_data)

pivot_df = df_vida.pivot_table(index='País', columns='Ano', values='Expectativa_Vida')

pivot_df['Média'] = pivot_df.mean(axis=1)
pivot_df = pivot_df.sort_values('Média', ascending=False).drop('Média', axis=1)

plt.figure(figsize=(14, 10), facecolor='white')

sns.heatmap(pivot_df, annot=True, fmt='.1f', cmap='RdYlGn_r',
            linewidths=0.5, linecolor='white', cbar_kws={'label': 'Expectativa de Vida (anos)'})

plt.title('Expectativa de Vida por País e Ano (2012-2021)',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Ano', fontsize=12, fontweight='bold')
plt.ylabel('País', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

##Scatterplot (gráfico de dispersão)
